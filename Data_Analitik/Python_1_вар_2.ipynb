## 4.Приоритет операций

Без четкого понимания приоритета операций вы рискуете получить неверные результаты. Python, как и любой калькулятор, выполняет операции в определенном порядке. Если вы не учитываете этот порядок, результат может быть не тем, который вы ожидали.

**Приоритет арифметических операций** (от высшего к низшему):

1.Скобки: () Выражения в скобках выполняются в первую очередь. Скобки можно использовать для изменения стандартного порядка выполнения операций.

2.Возведение в степень: ** (например, 2 ** 3 означает 2 в степени 3, что равно 8).

3.Унарные операторы: +, - (например, -5 - унарный минус, просто указывает на отрицательное число).

4.Умножение, деление, деление нацело, остаток от деления: *, /, //, % Эти операции имеют одинаковый приоритет и выполняются слева направо.
 * "*" (умножение)
 * / (деление с плавающей точкой)
 * // (деление нацело, результат - целое число)
 * % (остаток от деления)

5.Сложение и вычитание: +, - Эти операции имеют одинаковый приоритет и выполняются слева направо.

```python
# Пример 1: Умножение выполняется раньше сложения
result = 2 + 3 * 4      # 3 * 4 = 12, затем 2 + 12 = 14
print(result)  # Вывод: 14

# Пример 2: Скобки изменяют приоритет
result = (2 + 3) * 4          # 2 + 3 = 5, затем 5 * 4 = 20
print(result)  # Вывод: 20

# Пример 3: Возведение в степень выполняется раньше умножения
result = 2 * 3 ** 2         # 3 ** 2 = 9, затем 2 * 9 = 18
print(result)  # Вывод: 18

# Пример 4: Операции с одинаковым приоритетом выполняются слева направо
result = 10 / 2 * 3       # 10 / 2 = 5.0, затем 5.0 * 3 = 15.0
print(result)  # Вывод: 15.0

# Пример 5: Использование скобок для явного указания приоритета (повышает читаемость)
result = 10 / (2 * 3)      # 2 * 3 = 6, затем 10 / 6 = 1.6666666666666667
print(result)  # Вывод: 1.6666666666666667

# Пример 6: Деление нацело и остаток от деления
result = 17 // 5        # Результат деления 17 на 5 нацело (без остатка)
print(result) #Вывод: 3

result = 17 % 5 # Остаток от деления 17 на 5
print(result) # Вывод: 2
```

## 5.Преобразование между int и float

**Неявное преобразование.**

Любое целое число (тип int) можно использовать там, где ожидается число с плавающей точкой (тип float), поскольку при необходимости Python автоматически преобразует целые числа в числа с плавающей точкой.

**Явное преобразование.**

Число с плавающей точкой нельзя неявно преобразовать в целое число. Для такого преобразования необходимо использовать явное преобразование с помощью команды int().


```python
# Явное преобразование
num1 = 17.89
num2 = -13.56
num3 = int(num1)
num4 = int(num2)
print(num1)
print(num2)


```

Не путайте операцию преобразования и округления. Для округления чисел с плавающей точкой используются дополнительные команды

```python
# Неявное преобразование
num1 = 12  # был int
num2 = 5.3
num3 = float(num1) # станет float

print(num1+num2) # без перевода
print(num3+num2) # с переводом
```

 ## 6.Булевые переменные

Логический тип данных, или булев тип, или булевый тип, — тип данных в программировании, принимающий два возможных значения: истину и ложь.

```python
is_active = True  # правда
is_done = False   # ложь

print(type(is_active))
print(type(is_done))

print(is_active)
```

```python
print(3 > 5) #False
print(3 < 5) #True
print(3 == 5) #False
print(3 != 5) #True
print(3 >= 5) #False
print(3 <= 5) #True
```

Булевы значения могут использоваться в функциях. Некоторые функции возвращают значения True или False в качестве результата. Пример – **функция isinstance()**, с помощью которой можно определить, относится ли объект к указанному нами типу или нет.

```python
print(isinstance(25, int)) #True
print(isinstance(2.5, int)) #False
print(isinstance("Привет!", str)) #True

```

**NaN (Not a Number)**
Что это такое: **NaN**– это специальное значение в числовых типах данных (особенно в числах с плавающей точкой), которое представляет неопределенное, непредставимое или некорректное числовое значение. Оно возникает как результат математических операций, которые не могут дать осмысленный числовой результат.

NaN ведет себя очень специфично при сравнении: любое сравнение NaN с любым значением (включая само себя) дает False. Это связано с тем, что “Not a Number” по своей природе не является числом, и поэтому не может быть равно чему-либо.

### Задачка 4

Напишите программу, определяющую число десятков и единиц в двузначном числе.

Число единиц – это последняя цифра числа, число десятков – первая цифра.

Чтобы получить последнюю цифру любого числа, нужно найти остаток от деления числа на 10. Чтобы найти первую цифру двузначного числа, нужно поделить число нацело на 10. Программа, решающая поставленную задачу, может иметь следующий вид:

10

Число десятков = 1 Число единиц = 0

### ответ задачи 4

```python
num = int(input())      # получаем число пользователя
last_digit = num % 10    # получаем сдвиг  и берем   остаток от деления
first_digit = num // 10  #  получаем сдвиг берем десятки
print('Число десятков =', first_digit)
print('Число единиц =', last_digit)
```

задача 6

Расстояние между городами — 5900 километров.

Средняя скорость за весь путь будет 68 километров в час.

В день будете проезжать 4,5 часа.

Машина потребляет по 8,3 литра бензина на 100 километров.

цена на бензин — 65 рублей за литр.

Напишите код, который будет рассчитывать и выводить на экран количество полных дней путешествия (последний частичный день можно не считать), количество потраченного бензина, стоимость всего потраченного бензина.

### ответ задачи 6

```python
 # Сохраняем всю необходимую информацию из условия в переменные
distance = 5900
V = 68  # скорость км/ч
my_time = 4.5
potreblenie = 8.3 # бензин на 100 км
price =65
# Проводим необходимые вычисления
day = int(my_time * V)    # сколько в день
vsego_day = distance / day    #  кол-во дней на весь путь
bensin = distance * potreblenie / 100
# Выводим результаты на экран
print("Путешествие будет длиться", vsego_day ,"полных дней","количество потраченного бензина", bensin ,"стоимость всего потраченного бензина", bensin * price )
```
