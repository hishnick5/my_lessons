## **1.Визуализация Одномерных Распределений**

### Категориальные данные:

#### **sns.countplot()** -столбчатая диаграмма


функций Seaborn для категориальных данных. Она строит столбчатую диаграмму, где высота каждого столбца соответствует количеству наблюдений (частоте) в каждой категории.

**Основные параметры:**

* **x или y:** Имя столбца в DataFrame, который будет использоваться для группировки. Если указано x, категории будут по горизонтали. Если y, то по вертикали (удобно для длинных названий категорий).

* **data:** DataFrame, содержащий данные.

* **hue:** Имя столбца для дальнейшего разделения столбцов. Например, если вы строите countplot по полу (‘Sex’), и используете hue='Pclass', то для каждой категории пола будут отдельные столбцы для каждого класса кают.

* **order:** Список строк, задающий порядок отображения категорий на оси. По умолчанию категории сортируются по их частоте (от большего к меньшему).

* **hue_order**: Список строк, задающий порядок отображения категорий для hue.

* **color:** Единый цвет для всех столбцов.

* **palette:** Цветовая палитра для hue (например, ‘viridis’, ‘Blues’, [‘red’, ‘green’]).

* **orient:** Ориентация графика (‘v’ для вертикального, ‘h’ для горизонтального). Автоматически определяется по x и y.

* **saturation:** Интенсивность цвета (от 0 до 1).

##### пример на Titanic

```python
# 1. Количество пассажиров по классам кают
plt.figure(figsize=(8, 5))

sns.countplot(x='pclass', data=titanic)

plt.title('Количество пассажиров по классам кают')
plt.xlabel('Класс каюты')
plt.ylabel('Количество пассажиров')
plt.show()
```

```python
# 2. Количество пассажиров по полу
plt.figure(figsize=(6, 4))

sns.countplot(y='sex', data=titanic) # Используем y для лучшей читаемости

plt.title('Количество пассажиров по полу')
plt.xlabel('Количество пассажиров')
plt.ylabel('Пол')
plt.show()
```

```python
# 3. Количество пассажиров по классам кают, разделенное по полу (hue)
plt.figure(figsize=(8, 5))

sns.countplot(x='pclass', hue='sex', data=titanic, palette='viridis')

plt.title('Количество пассажиров по классам кают, разделенное по полу')
plt.xlabel('Класс каюты')
plt.ylabel('Количество пассажиров')
plt.show()
```

```python
# 4. Количество выживших (Survived=1) и погибших (Survived=0) по классам кают
plt.figure(figsize=(8, 5))

sns.countplot(x='pclass', hue='survived', data=titanic, palette='muted')

plt.title('Выживаемость пассажиров по классам кают')
plt.xlabel('Класс каюты')
plt.ylabel('Количество пассажиров')
# Изменим подписи легенды для ясности
plt.legend(title='Выжил', labels=['Нет', 'Да'])
plt.show()
```

```python
# 5. Изменение порядка категорий (например, если мы хотим показать классы в порядке 1, 2, 3)
plt.figure(figsize=(8, 5))

sns.countplot(x='pclass', data=titanic, order=[3, 2, 1])

plt.title('Количество пассажиров по классам кают (задан порядок)')
plt.xlabel('Класс каюты')
plt.ylabel('Количество пассажиров')
plt.show()
```

#### **sns.pieplot()**- Визуализация долей (как часть сектора)

**Важное замечание:** Seaborn не имеет отдельной, высокоуровневой функции pieplot(), как countplot() или scatterplot(). Традиционно круговые диаграммы (pie charts) создаются с помощью Matplotlib. Seaborn может быть использован для подготовки данных или для добавления стилей, но сам график рисуется через matplotlib.pyplot.pie().

Чаще всего мы используем Seaborn для агрегации данных, а затем передаем результат в Matplotlib. Например, мы можем использовать **value_counts()** или **groupby()** с **size()** для получения нужных данных.

##### пример на Titanic

```python
# Получим количество пассажиров для каждого класса кают
# value_counts() - это метод Pandas, который Seaborn часто использует "под капотом"
pclass_counts = titanic['pclass'].value_counts().sort_index()
```

```python
# Также можно получить через groupby:
# pclass_counts = titanic.groupby('pclass').size()

print("Количество пассажиров по классам кают:\n", pclass_counts)
```

```python
# --- Шаг 2: Создание круговой диаграммы с помощью Matplotlib ---
labels = pclass_counts.index.astype(str) + ' класс' # Метки для секторов
sizes = pclass_counts.values # Значения для секторов
```

```python
# Зададим цвета (можно использовать палитры Seaborn)

colors = sns.color_palette('cubehelix', len(labels))# Например, возьмем палитру 'viridis' и возьмем из нее 3 цвета

plt.figure(figsize=(7, 7)) # Размер фигуры для круговой диаграммы

# Создаем круговую диаграмму
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True, startangle=140)

# Добавляем заголовок
plt.title('Доля пассажиров по классам кают')
```

```python
# --- Еще один пример: Доля выживших ---
survival_counts = titanic['survived'].value_counts()
labels_survived = ['Погибли', 'Выжили'] # Более понятные метки
sizes_survived = survival_counts.values

colors_survived = sns.color_palette('pastel', len(labels_survived)) # Другая палитра

plt.figure(figsize=(6, 6))
plt.pie(sizes_survived, labels=labels_survived, colors=colors_survived, autopct='%1.1f%%',
        shadow=True, startangle=90, explode=(0.1, 0)) # Выдвинем первый сектор (погибли)

plt.title('Доля выживших и погибших на "Титанике"')
plt.axis('equal')
plt.show()
```

### Числовые данные

#### **sns.histplot()** - Гистограмма

**когда использовать**
* Показывает фактическое распределение данных в виде столбцов.
* Легко интерпретировать, особенно когда данных немного.
* Может быть зашумленной, если интервалов слишком много.
* Подходит для дискретных и непрерывных данных.

**Основные параметры:**

* **x или y**: Имя столбца в DataFrame, содержащего числовые данные.
* data: DataFrame, содержащий данные.
* bins:
 * Число: Количество интервалов (столбцов) в гистограмме. Большее количество интервалов дает более детальное представление о распределении, но может быть шумным. Меньшее количество интервалов дает более сглаженное представление.
 * Строка: Метод автоматического определения количества интервалов (например, ‘auto’, ‘fd’, ‘doane’, ‘scott’, ‘stone’, ‘rice’, ‘sturges’). Seaborn передает это аргумент в numpy.histogram_bin_edges.
 * Массив: Границы интервалов.
* kde: Булево значение. Если True, поверх гистограммы отображается оценка плотности ядра (KDE).
* stat:
 * ‘count’ (по умолчанию): Отображает количество наблюдений в каждом интервале.
 * ‘frequency’: Отображает частоту (долю) наблюдений в каждом интервале (суммируется в 1).
 * ‘density’: Нормализует столбцы так, чтобы общая площадь под гистограммой была равна 1. Это полезно для сравнения распределений с разным количеством наблюдений.
 * ‘probability’: Нормализует столбцы так, чтобы сумма их высот была равна 1.
* color: Цвет столбцов гистограммы.
* palette: Цветовая палитра для hue.
* hue: Имя столбца для разделения гистограммы по категориям.
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* multiple:
 * ‘layer’ (по умолчанию): Каждая категория hue отображается отдельно.
 * ‘stack’: Столбцы для разных категорий hue накладываются друг на друга.
 * ‘dodge’: Столбцы для разных категорий hue располагаются рядом друг с другом.
 * ‘fill’: Столбцы накладываются друг на друга и нормализуются так, чтобы общая высота составляла 1.
* element:
 * ‘bars’ (по умолчанию): Отображает столбцы.
 * ‘step’: Отображает контур столбцов.
 * ‘poly’: Заполняет область под контуром.
* fill: Булево значение. Заполнять ли столбцы цветом (только для element='step' или element='poly').
* shrink: Число от 0 до 1. Насколько сузить ширину столбцов.
* kde_kws: Словарь, содержащий параметры для sns.kdeplot(), если kde=True.


##### пример на датасете tips

```python
# 1. Простая гистограмма распределения общей суммы счета
plt.figure(figsize=(8, 5))

sns.histplot(x='total_bill', data=tips)

plt.title('Распределение общей суммы счета')
plt.xlabel('Сумма счета')
plt.ylabel('Количество')
plt.show()
```

```python
# 3. Гистограмма с оценкой плотности ядра (KDE)
plt.figure(figsize=(8, 5))

sns.histplot(x='total_bill', data=tips, kde=True)

plt.title('Распределение общей суммы счета с KDE')
plt.xlabel('Сумма счета')
plt.ylabel('Количество')
plt.show()
```

```python
# 4. Гистограмма с нормализацией (density)
plt.figure(figsize=(8, 5))

sns.histplot(x='total_bill', data=tips, stat='density')

plt.title('Распределение общей суммы счета (плотность)')
plt.xlabel('Сумма счета')
plt.ylabel('Плотность')
plt.show()
```

```python
# 5. Гистограмма с разделением по полу (hue)
plt.figure(figsize=(8, 5))

sns.histplot(x='total_bill', data=tips, hue='sex', multiple='stack')

plt.title('Распределение общей суммы счета по полу (stacked)')
plt.xlabel('Сумма счета')
plt.ylabel('Количество')
plt.show()
```

```python
# 6. Гистограмма с разделением по курящим (hue), столбцы рядом
plt.figure(figsize=(8, 5))

sns.histplot(x='total_bill', data=tips, hue='smoker', multiple='dodge')

plt.title('Распределение общей суммы счета по курящим (dodged)')
plt.xlabel('Сумма счета')
plt.ylabel('Количество')
plt.show()
```

```python
# 7. Гистограмма с другим стилем (step)
plt.figure(figsize=(8, 5))

sns.histplot(x='total_bill', data=tips, element='step', fill=True)

plt.title('Распределение общей суммы счета (step)')
plt.xlabel('Сумма счета')
plt.ylabel('Количество')
plt.show()
```

```python
# 8. Использование kde_kws для настройки KDE
plt.figure(figsize=(8, 5))

ax = sns.histplot(x='total_bill', data=tips, kde=True, color='skyblue') # Захватываем объект Axes

# Настраиваем толщину линии KDE
for line in ax.lines:
    line.set_linewidth(3)
    line.set_color('red')

plt.title('Распределение общей суммы счета с настроенным KDE')
plt.xlabel('Сумма счета')
plt.ylabel('Количество')
plt.show()
```

####  **sns.kdeplot()** - Оценка плотности ядра

KDE - это непараметрический способ оценки функции плотности вероятности случайной величины. По сути, это сглаженная версия гистограммы, которая показывает примерную форму распределения данных.

**когда использовать**
* Показывает оценку плотности распределения, сглаженную версию гистограммы.

* Помогает увидеть общую форму распределения, особенно когда данных много.
* Чувствительна к выбору ширины полосы (bw_adjust).
Лучше подходит для непрерывных данных.

**Основные параметры:**

* **x или y:** Имя столбца в DataFrame, содержащего числовые данные.

* data: DataFrame, содержащий данные.
* bw_adjust: Множитель для ширины полосы (bandwidth). Увеличение этого значения делает KDE более сглаженным.
* bw_method: Метод для определения ширины полосы (‘scott’, ‘silverman’, либо функция).
* fill: Булево значение. Заполнять ли область под кривой KDE цветом.
* color: Цвет кривой KDE.
* palette: Цветовая палитра для hue.
* hue: Имя столбца для разделения KDE по категориям.
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* multiple:
 * ‘layer’ (по умолчанию): Каждая категория hue отображается отдельно.
 * ‘stack’: Кривые KDE для разных категорий hue накладываются друг на друга.
 * ‘fill’: Кривые KDE накладываются друг на друга и нормализуются так, чтобы общая площадь под ними составляла 1.
* clip: Кортеж (min, max). Обрезает кривую KDE в указанном диапазоне.
* cumulative: Булево значение. Если True, отображает кумулятивную функцию распределения.
* common_norm: Булево значение. Если True, нормализует KDE для каждой категории hue относительно общего количества данных. Если False, каждая категория нормализуется отдельно.
* levels: Количество контурных линий для двумерного kdeplot.
* thresh: Значение между 0 и 1. Определяет уровень, ниже которого контурные линии не отображаются (двумерные kdeplot).

##### пример на датасете tips

```python
# 1. Простая KDE для общей суммы счета
plt.figure(figsize=(8, 5))

sns.kdeplot(x='total_bill', data=tips)

plt.title('KDE для общей суммы счета')
plt.xlabel('Сумма счета')
plt.ylabel('Плотность')
plt.show()
```

```python
# 2. KDE с заливкой
plt.figure(figsize=(8, 5))

sns.kdeplot(x='total_bill', data=tips, fill=True)

plt.title('KDE для общей суммы счета (с заливкой)')
plt.xlabel('Сумма счета')
plt.ylabel('Плотность')
plt.show()
```

```python
# 3. KDE с разделением по полу (hue)
plt.figure(figsize=(8, 5))

sns.kdeplot(x='total_bill', data=tips, hue='sex')

plt.title('KDE для общей суммы счета по полу')
plt.xlabel('Сумма счета')
plt.ylabel('Плотность')
plt.show()
```

```python
# 4. KDE с разделением по курящим (hue), stacked
plt.figure(figsize=(8, 5))

sns.kdeplot(x='total_bill', data=tips, hue='smoker', multiple='stack')

plt.title('KDE для общей суммы счета по курящим (stacked)')
plt.xlabel('Сумма счета')
plt.ylabel('Плотность')
plt.show()
```

```python
# 5. Изменение ширины полосы (bw_adjust)
plt.figure(figsize=(8, 5))

sns.kdeplot(x='total_bill', data=tips, bw_adjust=0.5)  # Меньше сглаживания

plt.title('KDE с bw_adjust=0.5')
plt.xlabel('Сумма счета')
plt.ylabel('Плотность')
plt.show()
```

```python
# 6. Кумулятивная KDE
plt.figure(figsize=(8, 5))

sns.kdeplot(x='total_bill', data=tips, cumulative=True)

plt.title('Кумулятивная KDE для общей суммы счета')
plt.xlabel('Сумма счета')
plt.ylabel('Кумулятивная плотность')
plt.show()
```

#### **sns.rugplot()** - Метки отдельных наблюдений

это функция Seaborn, которая добавляет маленькие черточки (метки) на оси графика, показывая расположение каждого отдельного наблюдения в одномерном распределении данных. Это помогает лучше понять плотность данных, особенно когда у вас есть информация о выбросах или кластеризации.

**Основные параметры:**

* x или y: Имя столбца в DataFrame, содержащего данные. Указывает ось, на которой будут отображаться черточки.
* data: DataFrame, содержащий данные.
* height: Высота черточек относительно оси. По умолчанию 0.025.
* color: Цвет черточек.
* ax: Объект matplotlib.axes.Axes, на котором будет нарисован график. Если не указан, создается новый.
* **kwargs: Дополнительные параметры для настройки черточек (например, linewidth, alpha).

##### пример на датасете tips

```python
# 1. Rugplot для total_bill
plt.figure(figsize=(8, 2)) # Уменьшаем высоту, чтобы лучше видеть черточки

sns.rugplot(x='total_bill', data=tips, color='green')

plt.title('Rugplot для общей суммы счета')
plt.xlabel('Сумма счета')
plt.show()
```

```python
# 2. Rugplot для tip, с изменением цвета и высоты
plt.figure(figsize=(8, 2))

sns.rugplot(x='tip', data=tips, color='blue', height=0.3) # Увеличиваем высоту черточек

plt.title('Rugplot для чаевых (tip)')
plt.xlabel('Размер чаевых')
plt.show()
```

```python
# 3. Совмещение rugplot с гистограммой для total_bill (для наглядности)
plt.figure(figsize=(8, 5))

sns.histplot(x='total_bill', data=tips, kde=False, bins=20, color='skyblue')
sns.rugplot(x='total_bill', data=tips, color='darkgreen', height=0.08)

plt.title('Гистограмма и Rugplot для общей суммы счета')
plt.xlabel('Сумма счета')
plt.ylabel('Количество')
plt.show()
```

```python
# 4. Rugplot для tip, с изменением цвета, высоты и kwargs
plt.figure(figsize=(8, 2))

sns.rugplot(x='tip', data=tips, color='blue', height=0.3,
            linewidth=1, alpha=0.3)  # Настройка linewidth и alpha

plt.title('Rugplot для чаевых (tip)')
plt.xlabel('Размер чаевых')
plt.show()
```

#### **sns.ecdfplot()** - Эмпирическая кумулятивная функция распределения (ECDF)

**ECDF** показывает, какую долю данных меньше или равна определенному значению. Это полезный инструмент для сравнения распределений, оценки перцентилей и понимания общего распределения данных.

1.Функция сортирует значения в столбце по возрастанию.

2.Для каждого значения она вычисляет долю всех значений, которые меньше или равны этому значению.

3.Затем она рисует график, где по оси X расположены значения данных, а по оси Y - соответствующая доля (от 0 до 1).

**Основные параметры:**

* x или y: Имя столбца в DataFrame, содержащего данные.

* data: DataFrame, содержащий данные.
* hue: Имя столбца для разделения ECDF по категориям.
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* stat: Статистика для построения ECDF. По умолчанию ‘proportion’ (пропорция).
* complementary: Булево значение. Если True, отображает complementary cumulative distribution  function (CCDF), т.е. 1 - ECDF (показывает, какую долю данных больше или равна определенному значению).
* fill: Булево значение. Заполнять ли область под кривой ECDF цветом.
* color: Цвет линии ECDF.
palette: Цветовая палитра для hue.
* line_kws: Словарь, содержащий параметры для настройки линий (например, linewidth, linestyle).
* marker: Символ маркера для точек данных.
* markers: Булево значение. Отображать ли маркеры на точках данных.
* **kwargs: Дополнительные параметры для настройки графика.

##### пример на на датасете tips

```python
# 1. ECDF для total_bill
plt.figure(figsize=(8, 5))

sns.ecdfplot(x='total_bill', data=tips, color='orange')

plt.title('ECDF для общей суммы счета')
plt.xlabel('Сумма счета')
plt.ylabel('Доля')
plt.show()
```

```python
# 2. ECDF для tip, с заливкой с ручным подсчетом
plt.figure(figsize=(8, 5))

# 1. Вычисляем ECDF вручную с помощью Pandas
sorted_tips = tips['tip'].sort_values()  # Сортируем значения
y_values = pd.Series(range(1, len(sorted_tips)+1)) / len(sorted_tips) # Вычисляем доли

# 2. Строим график с помощью Matplotlib
plt.plot(sorted_tips, y_values, color='purple', linewidth=2)

# 3. Заполняем область под графиком (опционально)
plt.fill_between(sorted_tips, y_values, color='purple', alpha=0.3)

plt.title('ECDF для размера чаевых (tip)')
plt.xlabel('Размер чаевых')
plt.ylabel('Доля')
plt.show()
```

```python
# 5. ECDF с маркерами и line_kws (для старых версий Seaborn - ИСПРАВЛЕНО)
plt.figure(figsize=(8, 5))

# 1. Вычисляем ECDF вручную с помощью Pandas
sorted_tips = tips['total_bill'].sort_values()  # Сортируем значения (для примера total_bill)
y_values = pd.Series(range(1, len(sorted_tips)+1)) / len(sorted_tips) # Вычисляем доли

# 2. Строим график с помощью Matplotlib
plt.plot(sorted_tips, y_values, linestyle='--', linewidth=2, color='darkgreen', marker='o', markerfacecolor='red', markersize=5)

plt.title('ECDF с маркерами и настройками линии')
plt.xlabel('Сумма счета')
plt.ylabel('Доля')
plt.show()
```

### **sns.scatterplot()** - Диаграмма рассеяния

предназначенная для визуализации взаимосвязи между двумя числовыми переменными. Она отображает данные в виде точек на плоскости, где положение каждой точки определяется значениями этих двух переменных. Диаграммы рассеяния позволяют выявлять тренды, кластеры, выбросы и другие закономерности в данных.

**Основные параметры:**

* x и y: Имена столбцов в DataFrame, определяющие координаты точек.

* data: DataFrame, содержащий данные.
* hue: Имя столбца для разделения точек по категориям (цвет точек).
* size: Имя столбца для разделения точек по размерам.
* style: Имя столбца для разделения точек по форме маркера.
* palette: Цветовая палитра для hue.
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* sizes: Список или кортеж, задающий минимальный и максимальный размер точек. Или словарь, сопоставляющий значения size с размерами точек.
* markers: Список или словарь, задающий форму маркеров для каждой категории style.
* alpha: Прозрачность точек (от 0 до 1).
* ax: Объект matplotlib.axes.Axes, на котором будет нарисован график.
* legend:
 * “auto” (по умолчанию),  
  * “brief” (показывать только подмножество элементов в легенде),

* x_jitter: Добавляет случайное смещение к координатам x точек. Полезно, когда точки сильно перекрываются.
* y_jitter: Добавляет случайное смещение к координатам y точек.
* s: Размер маркера в точках (устаревший способ, рекомендуется использовать size).

```python
import random  # Импортируем модуль random

# 5. Диаграмма рассеяния с прозрачностью (alpha) и случайным смещением (jitter)
plt.figure(figsize=(8, 6))

# Зададим величину смещения
jitter_amount = 0.5  # Регулируйте этот параметр по своему вкусу

# Добавим случайное смещение к координатам x и y
tips['total_bill_jittered'] = tips['total_bill'] + [random.uniform(-jitter_amount, jitter_amount) for _ in range(len(tips))]
tips['tip_jittered'] = tips['tip'] + [random.uniform(-jitter_amount, jitter_amount) for _ in range(len(tips))]

# Создадим диаграмму рассеяния с использованием новых столбцов
sns.scatterplot(x='total_bill_jittered', y='tip_jittered', data=tips, alpha=0.7, color='purple')

# Удалим временные столбцы (опционально)
tips.drop(['total_bill_jittered', 'tip_jittered'], axis=1, inplace=True)

plt.title('Диаграмма рассеяния: Сумма счета vs. Чаевые (с прозрачностью и смещением)')
plt.xlabel('Сумма счета')
plt.ylabel('Чаевые')
plt.show()
```

### **sns.lineplot()**- Линейный график

используется для отображения взаимосвязи между двумя числовыми переменными в виде линии. Этот тип графика особенно полезен для визуализации трендов, изменений во времени, или когда данные упорядочены каким-либо другим образом.

**Важное замечание:**

sns.lineplot() предполагает, что данные упорядочены по оси X.

Если это не так, результаты могут быть неверными. Перед использованием **sns.lineplot()** убедитесь, что данные отсортированы по переменной, отображаемой на оси X.

 Если ваши данные не отсортированы, **sns.scatterplot()** будет более подходящим выбором.

**Основные параметры:**

* x и y: Имена столбцов в DataFrame, определяющие координаты точек линии.

* data: DataFrame, содержащий данные.
* hue: Имя столбца для разделения линий по категориям (цвет линий).
* size: Имя столбца для разделения линий по толщине.
* style: Имя столбца для разделения линий по типу линии.
* palette: Цветовая палитра для hue.
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* sizes: Список или кортеж, задающий минимальную и максимальную толщину линий. Или словарь, сопоставляющий значения size с толщиной линий.
* dashes: Список или словарь, задающий тип штриховки для каждой категории style.
* markers: Булево значение. Отображать ли маркеры на точках линии.
* err_style: Стиль отображения доверительного интервала (‘band’ или ‘bars’).
* err_kws: Словарь, содержащий параметры для настройки доверительного интервала (например, alpha, color).
* estimator: Функция для агрегирования данных (по умолчанию mean, можно использовать sum, median, None и т.д.). Если estimator=None, отображаются все точки данных, а не агрегированные значения.
* units: Имя столбца, определяющего независимые единицы измерения. Используется для расчета доверительных интервалов.
* ax: Объект matplotlib.axes.Axes, на котором будет нарисован график.
* legend: “auto” (по умолчанию), “brief”, “full”, False.

### **sns.jointplot()** - Объединенные графики

это очень удобная функция, которая создает объединенные графики. Она позволяет визуализировать взаимосвязь между двумя переменными, а также их индивидуальные распределения на одном графике. Основной график в центре - это диаграмма рассеяния, а по краям (на полях) отображаются одномерные графики (например, гистограммы или KDE) для каждой переменной.

**Основные параметры:**

* x и y: Имена столбцов в DataFrame, определяющие координаты точек.

* data: DataFrame, содержащий данные.
* kind: Тип графика, который будет использоваться:
 * 'scatter' (по умолчанию): Диаграмма рассеяния в центре, гистограммы по краям.
 * 'kde': Двумерная оценка плотности (KDE) в центре, одномерные KDE по краям.
 * 'hist': Двумерная гистограмма (heatmap) в центре, гистограммы по краям.
 * 'reg': Диаграмма рассеяния с регрессионной линией.
 * 'resid': Диаграмма рассеяния с остатками регрессии.
* color: Цвет для всех элементов графика.
* palette: Цветовая палитра для hue.
* hue: Имя столбца для разделения точек по категориям (цвет точек).
* height: Размер (ширина и высота) всего графика в дюймах.
* ratio: Соотношение размеров центральной части графика к полям.
* space: Расстояние между центральной частью и полями.
* marginal_kws: Словарь с параметрами для одномерных графиков на полях (например, bins, color, kde, stat).
* joint_kws: Словарь с параметрами для центрального графика.
* annot_kws: Словарь с параметрами для аннотаций (например, для отображения R-квадрата при kind='reg').

##### примеры на датасете tips

```python
# 1. Jointplot с kind='scatter' (по умолчанию)
plt.figure(figsize=(8, 8)) # Фигура не нужна, jointplot сам управляет

g = sns.jointplot(x='total_bill', y='tip', data=tips, color='teal')
g.set_axis_labels('Сумма счета', 'Чаевые')

plt.suptitle('Jointplot: Scatterplot + Histograms', y=1.02) # y=1.02 поднимает заголовок
plt.show()
```

```python
# 2. Jointplot с kind='kde' и яркими цветами
plt.figure(figsize=(8, 8))

g = sns.jointplot(x='total_bill', y='tip', data=tips, kind='kde')
g.set_axis_labels('Сумма счета', 'Чаевые')

plt.suptitle('Jointplot: KDE', y=1.02)
plt.show()
```

```python
# 3. Jointplot с kind='kde' и разделением по полу (hue)
plt.figure(figsize=(8, 8))

g = sns.jointplot(x='total_bill', y='tip', data=tips, hue='sex', kind='kde', palette=['lime', 'violet'])
g.set_axis_labels('Сумма счета', 'Чаевые')

plt.suptitle('Jointplot: KDE по полу', y=1.02)
plt.show()
```

```python
# 4. Jointplot с kind='hist' и настройками полей
plt.figure(figsize=(8, 8))

g = sns.jointplot(x='total_bill', y='tip', data=tips, kind='hist',color='purple',
                  marginal_kws={'bins': 50, 'color': 'orange'})
g.set_axis_labels('Сумма счета', 'Чаевые')

plt.suptitle('Jointplot: Histograms', y=1.02)
plt.show()
```

```python
# 5. Jointplot с kind='reg'
plt.figure(figsize=(8, 8))

g = sns.jointplot(x='total_bill', y='tip', data=tips, kind='reg', color='darkblue',
                              marginal_kws={'bins': 50, 'color': 'blue'})

g.set_axis_labels('Сумма счета', 'Чаевые')

plt.suptitle('Jointplot: Regression', y=1.02)
plt.show()
```

### **sns.kdeplot()** - Двумерная оценка плотности

может быть использован не только для одномерных распределений, но и для двумерных. Двумерная KDE показывает плотность распределения совместной выборки двух переменных. Это выглядит как “тепловая карта” или контурные линии, показывающие области с наибольшей концентрацией точек.

**Основные параметры (для двумерного KDE):**

* x и y: Имена столбцов в DataFrame, определяющие переменные для осей.
* data: DataFrame, содержащий данные.
* fill: Булево значение. Заполнять ли области между контурными линиями цветом.
* color: Цвет контурных линий или заливки.
* palette: Цветовая палитра для hue.
* hue: Имя столбца для разделения KDE по категориям.
* levels: Количество контурных линий.
* thresh: Значение между 0 и 1. Определяет уровень, ниже которого контурные линии не отображаются (полезно для очистки графика).
* alpha: Прозрачность заливки.
* cmap: Цветовая карта для заливки (если fill=True).
* bw_adjust: Множитель для ширины полосы (bandwidth) при расчете KDE.
* ax: Объект matplotlib.axes.Axes, на котором будет нарисован график.

```python
# 1. Двумерный KDE для total_bill и tip
plt.figure(figsize=(8, 6))

ax = sns.kdeplot(x='total_bill', y='tip', data=tips, color='indigo', levels=5) #  указываем количество уровней -1

# Настраиваем ширину линии
for i, line in enumerate(ax.collections):
  line.set_linewidth(2)

plt.title('Двумерный KDE: Сумма счета vs. Чаевые')
plt.xlabel('Сумма счета')
plt.ylabel('Чаевые')
plt.show()
```

```python
# 2. Двумерный KDE с заливкой и яркой цветовой картой
plt.figure(figsize=(8, 6))

sns.kdeplot(x='total_bill', y='tip', data=tips, fill=True, cmap='viridis', alpha=0.5)

plt.title('Двумерный KDE с заливкой')
plt.xlabel('Сумма счета')
plt.ylabel('Чаевые')
plt.show()
```

### sns.pairplot(): Матрица диаграмм рассеяния и гистограмм

это мощная и удобная функция, которая создает матрицу попарных графиков для всех (или выбранных) числовых переменных в DataFrame. Она позволяет быстро получить общее представление о взаимосвязях между переменными и их распределениях.

**Важно:**

sns.pairplot() может занять много времени и ресурсов при работе с большим количеством переменных и большим объемом данных.

Помните, что sns.pairplot() предназначен в основном для числовых переменных.

 Использование категориальных переменных может привести к неожиданным результатам.

plot_kws и diag_kws позволяют очень тонко настраивать внешний вид графиков.


**Основные параметры:**

* data: DataFrame, содержащий данные.

* hue: Имя столбца для разделения графиков по категориям (цвет точек/линий/распределений).
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* palette: Цветовая палитра для hue.
* vars: Список имен столбцов для отображения в матрице. Если не указан, используются все числовые столбцы.
* kind: Тип графиков, которые будут использоваться вне диагонали:
 * 'scatter' (по умолчанию): Диаграммы рассеяния.
 * 'kde': Двумерные KDE.
 * 'hist': Двумерные гистограммы.
 * 'reg': Диаграммы рассеяния с регрессионной линией.
* diag_kind: Тип графика для отображения на диагонали:
 * 'hist' (по умолчанию): Гистограммы.
 * 'kde': KDE.
* markers: Список или словарь, задающий форму маркеров для каждой категории hue.
* plot_kws: Словарь с параметрами, которые будут переданы графикам вне диагонали (например, alpha, s для kind='scatter', levels для kind='kde').
* diag_kws: Словарь с параметрами, которые будут переданы графикам на диагонали (например, bins, kde для  diag_kind='hist', shade для diag_kind='kde').
* corner: Булево значение. Если True, отображает только нижнюю половину матрицы (без дублирующихся графиков).
* dropna: Булево значение. Если True, удаляет строки с отсутствующими значениями перед построением графиков.
* layout_pad: Отступ между подграфиками.
* height: Высота каждого графика в дюймах.
* aspect: Соотношение сторон каждого графика (ширина/высота).
* size: Размер каждого графика (в дюймах). Устаревший параметр, рекомендуется использовать height и aspect.

##### примеры на датасете tips

```python
# 1. Базовый pairplot
plt.figure(figsize=(10, 10))  # figure не нужен, pairplot сам создает

sns.pairplot(tips) # расчет по числовым данным

plt.suptitle('Pairplot: Basic', y=1.02)
plt.show()
```

```python
# 2. Pairplot с разделением по полу (hue) и яркой палитрой
plt.figure(figsize=(10, 10))

sns.pairplot(tips, hue='sex', palette=['darkorange', 'darkviolet'])

plt.suptitle('Pairplot: по полу (hue)', y=1.02)
plt.show()
```

```python
# 3. Pairplot с kde на диагонали и настройкой графика
plt.figure(figsize=(10, 10))

sns.pairplot(tips, diag_kind='kde', plot_kws={'alpha': 0.6}, diag_kws={'fill': True, 'color': 'green'})

plt.suptitle('Pairplot: kde diag', y=1.02)
plt.show()
```

```python
# 4. Pairplot с выбором переменных (vars) и настройкой диагональных гистограмм
plt.figure(figsize=(10, 10))
# какие именно столбцы использовать
sns.pairplot(tips, vars=['total_bill', 'tip'], diag_kind='hist', diag_kws={'color': 'forestgreen'})

plt.suptitle('Pairplot: Выбор переменных', y=1.02)
plt.show()
```

```python
# 5. Pairplot с разделением по полу (hue), разными маркерами и угловой версией
plt.figure(figsize=(10, 10))

sns.pairplot(tips, hue='sex', palette=['teal', 'salmon'], markers=['o', 's'], corner=True)

plt.suptitle('Pairplot: corner версия', y=1.02)
plt.show()
```

**Ключевые моменты и лучшие практики:**

* Начните с базового графика: Сначала создайте простой sns.pairplot(data=your_data), чтобы получить общее представление.
* Используйте hue: Если у вас есть категориальная переменная, используйте hue для разделения графиков по категориям и выявления различий.
* Выбирайте переменные: Если у вас много столбцов, используйте параметр vars, чтобы отобразить только наиболее интересные переменные.
* Настраивайте графики: Используйте plot_kws и diag_kws для настройки внешнего вида графиков (прозрачность, цвет, форма маркеров, количество бинов и т.д.).
* Помните об ограничениях: sns.pairplot() хорошо работает с небольшим и средним количеством числовых переменных. Для больших наборов данных рассмотрите другие методы визуализации или выборку данных.
* Не забудьте про plt.suptitle(): Добавьте общий заголовок с помощью plt.suptitle(), чтобы график был более понятным.

### **sns.boxplot()**- Диаграмма “ящик с усами”

это стандартный способ визуализации распределения данных на основе пятичисловой сводки: минимальное значение, первый квартиль (25-й перцентиль), медиана (50-й перцентиль), третий квартиль (75-й перцентиль) и максимальное значение. “Ящик” представляет интерквартильный размах (IQR), а “усы” простираются до минимального и максимального значений в пределах 1.5 * IQR. Значения за пределами “усов” считаются выбросами и отображаются в виде отдельных точек.

**Основные параметры:**


* x или y: Имя столбца в DataFrame для отображения. Обычно используется одна числовая переменная (для отображения распределения) и одна категориальная (для разделения ящиков по категориям).
* data: DataFrame, содержащий данные.
* hue: Имя столбца для разделения ящиков по категориям (цвет ящиков).
* order: Список строк, задающий порядок отображения категорий для x.
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* palette: Цветовая палитра для hue.
* orient: Ориентация ящиков (‘v’ - вертикальная, по умолчанию, ‘h’ - горизонтальная).
* width: Ширина ящиков (относительно расстояния между ними).
* color: Цвет для всех ящиков (если hue не используется).
* linewidth: Толщина линий ящиков и усов.
* fliersize: Размер точек, обозначающих выбросы.
* flierprops: Словарь с параметрами для настройки внешнего вида выбросов (например, marker, markerfacecolor, markeredgecolor).
* ax: Объект matplotlib.axes.Axes, на котором будет нарисован график.

##### примеры на датасете tips

```python
# 1. Базовый boxplot для total_bill
plt.figure(figsize=(8, 6))

sns.boxplot(x=tips['total_bill'], color='skyblue')

plt.title('Boxplot: Общая сумма счета')
plt.xlabel('Сумма счета')
plt.show()
```

```python
# 2. Boxplot для total_bill, разделенный по дням недели (x и y)
plt.figure(figsize=(8, 6))

sns.boxplot(x='day', y='total_bill', data=tips, palette='pastel') # Pastel-мягкие пастельные цвета

plt.title('Boxplot: Сумма счета по дням недели')
plt.xlabel('День недели')
plt.ylabel('Сумма счета')
plt.show()
```

```python
# 3. Boxplot с разделением по полу (hue) и яркой палитрой
plt.figure(figsize=(8, 6))

sns.boxplot(x='day', y='total_bill', data=tips, hue='sex', palette=['springgreen', 'darkorange'])

plt.title('Boxplot: Сумма счета по дням недели (по полу)')
plt.xlabel('День недели')
plt.ylabel('Сумма счета')
plt.show()
```

```python
# 4. Boxplot с горизонтальной ориентацией и настройкой выбросов
plt.figure(figsize=(8, 6))

sns.boxplot(y='day', x='total_bill', data=tips, orient='h',
            flierprops={'marker': 'd', 'markerfacecolor': '0.75', 'markersize': 5, 'markeredgecolor': 'red'}) #ромбики для выбросов

plt.title('Boxplot: Сумма счета по дням недели (горизонтальный, настройка выбросов)')
plt.ylabel('День недели')
plt.xlabel('Сумма счета')
plt.show()
```

### **sns.violinplot()** - Диаграмма скрипки (Violin Plot)

Предоставляет более подробную информацию о распределении данных. Вместо простого “ящика” он отображает оценку плотности вероятности (KDE) распределения, которая симметрично располагается по обе стороны от центральной линии. Форма “скрипки” показывает, где данные более плотные. Внутри “скрипки” может быть отображен boxplot (по умолчанию).

**Основные параметры:**

* x или y: Имя столбца в DataFrame для отображения.

* data: DataFrame, содержащий данные.
* hue: Имя столбца для разделения скрипок по категориям (цвет скрипок).
* order: Список строк, задающий порядок отображения категорий для x.
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* palette: Цветовая палитра для hue.
* orient: Ориентация скрипок (‘v’ - вертикальная, по умолчанию, ‘h’ - горизонтальная).
* width: Ширина скрипок (относительно расстояния между ними).
* color: Цвет для всех скрипок (если hue не используется).
* inner: Тип отображения внутри скрипки:
 * 'box': Boxplot (по умолчанию).
 * 'quartiles': Линии, показывающие квартили.
 * 'point': Точки, показывающие отдельные наблюдения.
 * 'stick': Маленькие черточки, показывающие отдельные наблюдения.
 * None: Ничего не отображать внутри скрипки.
* split: Булево значение. Если True и hue имеет только два значения, разделяет скрипку на две части, показывая распределение для каждой категории hue с разных сторон.
* scale: Метод масштабирования ширины скрипки:
 * 'area' (по умолчанию): Все скрипки имеют одинаковую площадь.
 * 'count': Ширина скрипки пропорциональна количеству наблюдений.
 * 'width': Все скрипки имеют одинаковую ширину.
* scale_hue: Булево значение. При scale='count' и наличии hue, масштабировать ли ширину скрипок в зависимости от количества наблюдений в каждой категории hue.
* bw: Ширина полосы (bandwidth) для оценки плотности KDE.
* ax: Объект matplotlib.axes.Axes, на котором будет нарисован график.

### **sns.stripplot()** - Точечная диаграмма

это простая диаграмма рассеяния, где одна переменная является категориальной. Она отображает все точки данных вдоль оси категориальной переменной. Это полезно для визуализации распределения данных и выявления отдельных значений в каждой категории. Однако, когда данных много, точки могут сильно перекрываться, затрудняя чтение графика.

**Основные параметры:**

* x или y: Имя столбца в DataFrame для отображения. Обычно используется одна числовая переменная и одна категориальная.
* data: DataFrame, содержащий данные.
* hue: Имя столбца для разделения точек по категориям (цвет точек).
* order: Список строк, задающий порядок отображения категорий для x.
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* jitter: Величина случайного смещения точек вдоль оси категориальной переменной. Это помогает уменьшить перекрытие точек и сделать график более читаемым. Значение 0 (по умолчанию) отключает смещение.
* dodge: Булево значение. Если True и используется hue, точки для каждой категории hue будут смещены рядом друг с другом.
* palette: Цветовая палитра для hue.
* orient: Ориентация точек (‘v’ - вертикальная, по умолчанию, ‘h’ - горизонтальная).
* color: Цвет для всех точек (если hue не используется).
* size: Размер точек.
* linewidth: Толщина линий контура точек.
* edgecolor: Цвет контура точек.
* alpha: Прозрачность точек (от 0 до 1).
* ax: Объект matplotlib.axes.Axes, на котором будет нарисован график.

##### Примеры на датасете tips

```python
# 3. Stripplot с разделением по полу (hue) и dodge=True
plt.figure(figsize=(8, 6))

sns.stripplot(x='day', y='total_bill', data=tips, hue='sex', dodge=True, palette=['limegreen', 'darkorange'])

plt.title('Stripplot: С разделением по полу и dodge')
plt.xlabel('День недели')
plt.ylabel('Сумма счета')
plt.show()
```

```python
# 4. Stripplot с настройкой размера и прозрачности точек
plt.figure(figsize=(8, 6))

sns.stripplot(x='day', y='total_bill', data=tips, size=7, alpha=0.4, color='purple')

plt.title('Stripplot: Настройка размера и прозрачности')
plt.xlabel('День недели')
plt.ylabel('Сумма счета')
plt.show()
```

### **sns.swarmplot()** - Диаграмма роя

Он располагает точки таким образом, чтобы они не накладывались друг на друга, что позволяет лучше визуализировать распределение данных и количество точек в каждой категории. Это особенно полезно, когда данных много и перекрытие является проблемой.

**Важное отличие:**

* data: DataFrame, содержащий данные.
* hue: Имя столбца для разделения точек по категориям (цвет точек).

* hue_order: Список строк, задающий порядок отображения категорий для hue.
* dodge: Булево значение. Если True и используется hue, точки для каждой категории hue будут смещены рядом друг с другом.
* palette: Цветовая палитра для hue.
* orient: Ориентация точек (‘v’ - вертикальная, по умолчанию, ‘h’ - горизонтальная).

### **sns.catplot()** - Универсальный график категориальных данных

это очень мощная и гибкая функция, которая позволяет создавать различные типы графиков для визуализации отношений, в которых участвует хотя бы одна категориальная переменная.

 Она объединяет в себе функциональность нескольких других функций Seaborn, таких как sns.barplot(), sns.violinplot(), sns.boxplot(), sns.stripplot(), sns.swarmplot() и sns.pointplot(), и предоставляет унифицированный интерфейс для их использования.

  Кроме того, sns.catplot() умеет создавать несколько графиков (facets) на основе значений других категориальных переменных, что делает ее особенно полезной для анализа сложных данных.

**Ключевые особенности:**

Тип графика: Вы можете выбрать тип графика, который будет отображаться, используя параметр kind.

Faceting: sns.catplot() может автоматически создавать несколько графиков, разделенных по значениям одной или нескольких категориальных переменных. Это называется “faceting”.

Figure-level function: sns.catplot() - это функция “уровня фигуры” (figure-level function), что означает, что она создает свою собственную фигуру Matplotlib и управляет ее структурой.

 В отличие от функций “уровня осей” (axes-level functions), таких как sns.scatterplot() или sns.histplot(), которые рисуют графики на существующих осях.

Основные параметры:

* x или y: Имя столбца в DataFrame для отображения (обычно категориальная переменная).

* data: DataFrame, содержащий данные.
* hue: Имя столбца для разделения графиков по категориям (цвет и, возможно, другие визуальные атрибуты).
* row: Имя столбца для разделения графиков по строкам (faceting).
* col: Имя столбца для разделения графиков по столбцам (faceting).
* col_wrap: Если col указан, то сколько графиков должно быть в одной строке. Полезно, когда много категорий в col.
* kind: Тип графика для отображения:
 * 'strip' (по умолчанию): sns.stripplot()
 * 'swarm': sns.swarmplot()
 * 'box': sns.boxplot()
 * 'violin': sns.violinplot()
 * 'bar': sns.barplot()
 * 'count': sns.countplot() (показывает количество наблюдений в каждой категории)
 * 'point': sns.pointplot()
* order: Список строк, задающий порядок отображения категорий для x.
* hue_order: Список строк, задающий порядок отображения категорий для hue.
* palette: Цветовая палитра для hue.
* estimator: Функция для агрегирования данных (используется для kind='bar' и kind='point').
* ci: Размер доверительного интервала (используется для kind='bar' и kind='point').
* sharex: Разделять ли оси X между графиками (по умолчанию True).
* sharey: Разделять ли оси Y между графиками (по умолчанию True).
* height: Высота каждого отдельного графика в дюймах.
* aspect: Соотношение сторон каждого отдельного графика (ширина/высота).
* col_order: Порядок отображения колонок
* row_order: Порядок отображения строк


**Важно:**

sns.catplot() возвращает объект FacetGrid, который позволяет настраивать отдельные графики и всю фигуру.

Поскольку sns.catplot() - это функция “уровня фигуры”, она создает собственную фигуру Matplotlib.

Не нужно создавать plt.figure() перед вызовом sns.catplot().

Для доступа к отдельным осям в FacetGrid, используйте методы map() или axes.


##### Примеры на датасете tips

```python
# Catplot с kind='strip'

g = sns.catplot(x='day', y='total_bill', data=tips, kind='strip', color='skyblue')
g.fig.suptitle('Catplot: Stripplot', y=1.03)

plt.show()
```

```python
#  Catplot с kind='count'

g = sns.catplot(x='day', data=tips, kind='count', palette='muted')
g.fig.suptitle('Catplot: Countplot', y=1.03)

plt.show()
```

```python
# Catplot с kind='swarm' и разделением по полу (hue)
g = sns.catplot(x='day', y='total_bill', data=tips, hue='sex', kind='swarm', palette=['limegreen', 'darkorange'])
g.fig.suptitle('Catplot: Swarmplot с hue', y=1.03)

plt.show()
```

```python
# 3. Catplot с kind='box' и разделением по курящим (col)

g = sns.catplot(x='day', y='total_bill', data=tips, col='smoker', kind='box', palette='pastel')
g.fig.suptitle('Catplot: Boxplot с col', y=1.03)

plt.show()
```

### **sns.regplot()** - Диаграмма рассеяния с регрессионной линией

это функция для отображения связи между двумя числовыми переменными, с добавлением регрессионной линии и доверительного интервала. Она создает диаграмму рассеяния точек данных и накладывает на нее линию, которая наилучшим образом описывает тренд в данных. Доверительный интервал показывает диапазон значений, в котором, вероятно, будет находиться истинная регрессионная линия.

**Основные параметры:**

* x и y: Имена столбцов в DataFrame, определяющие переменные для осей.
* data: DataFrame, содержащий данные.
* x_estimator: Функция для оценки центрального значения для каждого уникального значения x. По умолчанию None.
* x_bins: Если указано, данные будут сгруппированы по значениям x в указанное количество бинов, и для каждого бина будет рассчитано среднее значение y.
* x_ci: Размер доверительного интервала для x_estimator (от 0 до 100) или 'sd' (стандартное отклонение) или None (не отображать доверительный интервал).
* fit_reg: Булево значение. Отображать ли регрессионную линию и доверительный интервал (по умолчанию True).
* ci: Размер доверительного интервала (от 0 до 100).
* n_boot: Количество итераций бутстрапа для расчета доверительного интервала.
* order: Степень полиномиальной регрессии. order=1 (по умолчанию) - линейная регрессия. order=2 - квадратичная регрессия, и т.д.
* robust: Булево значение. Использовать ли робастную регрессию, которая менее чувствительна к выбросам.
* truncate: Булево значение. Обрезать ли регрессионную линию, чтобы она не выходила за пределы диапазона данных.
* x_partial и y_partial: Матрицы или массивы, содержащие значения дополнительных переменных, которые нужно учесть при расчете регрессии.
* scatter: Булево значение. Отображать ли точки данных.
* fit_kws: Словарь с параметрами, которые будут переданы функции statsmodels.OLS (используется для настройки регрессии).
* scatter_kws: Словарь с параметрами, которые будут переданы функции plt.scatter() (используется для настройки точек).
* line_kws: Словарь с параметрами, которые будут переданы функции plt.plot() (используется для настройки регрессионной линии).
* color: Цвет для точек и линии (если не указаны scatter_kws и line_kws).
* marker: Тип маркера для точек.
* ax: Объект matplotlib.axes.Axes, на котором будет нарисован график.

##### Примеры на датасете tips

```python
# 1. Базовый regplot для total_bill и tip
plt.figure(figsize=(8, 6))

sns.regplot(x='total_bill', y='tip', data=tips, color='skyblue')
plt.title('Regplot: Сумма счета vs. Чаевые')

plt.xlabel('Сумма счета')
plt.ylabel('Чаевые')
plt.show()
```

```python
# 2. Regplot с отключением отображения точек (scatter=False)
plt.figure(figsize=(8, 6))

sns.regplot(x='total_bill', y='tip', data=tips, scatter=False, color='coral')

plt.title('Regplot: Без точек')
plt.xlabel('Сумма счета')
plt.ylabel('Чаевые')
plt.show()
```

```python
# 3. Regplot с полиномиальной регрессией (order=2)
plt.figure(figsize=(8, 6))

sns.regplot(x='total_bill', y='tip', data=tips, order=2, color='limegreen')

plt.title('Regplot: Полиномиальная регрессия (order=2)')
plt.xlabel('Сумма счета')
plt.ylabel('Чаевые')
plt.show()
```

```python
# 4. Regplot с робастной регрессией (robust=True)
plt.figure(figsize=(8, 6))

sns.regplot(x='total_bill', y='tip', data=tips, robust=True, color='darkorange')

plt.title('Regplot: Робастная регрессия')
plt.xlabel('Сумма счета')
plt.ylabel('Чаевые')
plt.show()
```

### **sns.lmplot()** - Диаграмма рассеяния с регрессионной линией и Faceting

 это функция уровня фигуры (figure-level function) (как и catplot), которая объединяет в себе функциональность sns.regplot() и возможность создания нескольких графиков (faceting) на основе значений одной или нескольких категориальных переменных. Она создает диаграмму рассеяния с регрессионной линией и доверительным интервалом, и может автоматически создавать несколько таких графиков, разделенных по строкам и/или столбцам.

**Важно:**

* sns.lmplot() возвращает объект FacetGrid, который позволяет настраивать отдельные графики и всю фигуру.

* Поскольку sns.lmplot() - это функция “уровня фигуры”, она создает свою собственную фигуру Matplotlib. Не нужно создавать plt.figure() перед вызовом sns.lmplot().

* sns.lmplot хорошо работает когда регрессия линейна, в случаях не линейности лучше использовать sns.catplot и sns.regplot


**Основные параметры:**

Параметры sns.lmplot() включают все параметры sns.regplot(), а также параметры для faceting:

* aspect: Соотношение сторон каждого отдельного графика (ширина/высота).
* sharex: Разделять ли оси X между графиками (по умолчанию True).
* sharey: Разделять ли оси Y между графиками (по умолчанию True).
* legend: Булево значение. Отображать ли легенду (по умолчанию True).
* legend_out: Булево значение. Выводить ли легенду за пределы графиков.

### Настройка регрессионной модели (order, logistic, lowess)

**1. order - Степень полиномиальной регрессии**

Параметр order определяет степень полинома, используемого для построения регрессионной линии. По умолчанию (order=1), используется линейная регрессия. Если order=2, то используется квадратичная регрессия, order=3 - кубическая, и так далее.

Линейная регрессия (order=1): Подходит, когда между переменными существует линейная зависимость. Регрессионная линия будет прямой.

Полиномиальная регрессия (order > 1): Подходит, когда между переменными существует нелинейная зависимость. Регрессионная линия будет кривой. Увеличение order позволяет моделировать более сложные кривые, но может привести к переобучению (модель слишком хорошо подстраивается под обучающие данные и плохо обобщается на новые данные).

**Ключевые моменты и рекомендации:**

Выбор order: Начните с order=1 (линейная регрессия) и увеличьте order, только если между переменными явно нелинейная зависимость. Следите за переобучением.

logistic: Используйте logistic=True, когда переменная y является бинарной.

lowess: Используйте lowess=True, когда между переменными существует сложная нелинейная зависимость, и вы не хотите задавать конкретную форму функции.

LOWESS может быть вычислительно дорогим для больших наборов данных.

Всегда визуально анализируйте полученные графики, чтобы убедиться, что выбранная модель адекватно описывает данные.


```python
# Сгенерируем искусственные данные с нелинейной зависимостью
rng = np.random.default_rng(seed=42)
x = np.linspace(0, 10, 100)
y = x**2 + rng.normal(0, 10, 100)
```

```python
# 1. Линейная регрессия
plt.figure(figsize=(8, 6))
sns.regplot(x=x, y=y, order=1, label='Линейная регрессия', color='skyblue')
plt.title('Регрессия: Линейная')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
# 2. Квадратичная регрессия
sns.regplot(x=x, y=y, order=2, label='Квадратичная регрессия', color='coral')
plt.title('Регрессия: Квадратичная')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
# 3. Кубическая регрессия
sns.regplot(x=x, y=y, order=3, label='Кубическая регрессия', color='limegreen')
plt.title('Регрессия: Кубическая')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

**2. logistic - Логистическая регрессия**

Параметр logistic используется, когда переменная y является бинарной (принимает только два значения, например, 0 и 1). Логистическая регрессия моделирует вероятность того, что y равно 1, в зависимости от значения x.

 Она отображает S-образную кривую (логистическую функцию), которая показывает, как вероятность изменяется с изменением x. Параметр logistic должен быть установлен в True.

```python
# Сгенерируем искусственные данные для логистической регрессии
rng = np.random.default_rng(seed=42)
x = np.linspace(-5, 5, 100)
p = 1 / (1 + np.exp(-x))  # Логистическая функция
y = rng.binomial(1, p, 100) # Бинарная переменная (0 или 1)
```

```python
# Логистическая регрессия
plt.figure(figsize=(8, 6))
sns.regplot(x=x, y=y, logistic=True, color='skyblue')
plt.title('Логистическая регрессия')
plt.xlabel('x')
plt.ylabel('Вероятность (y=1)')
plt.show()
```

**lowess - Локально взвешенная полиномиальная регрессия**

LOWESS (Locally Weighted Scatterplot Smoothing) - это непараметрический метод регрессии, который не предполагает конкретную форму функции, связывающей x и y. Он использует локальную регрессию для каждой точки данных, взвешивая соседние точки в зависимости от их расстояния до текущей точки. Это позволяет моделировать сложные нелинейные зависимости, не задавая глобальную функцию.

Чтобы использовать LOWESS в sns.regplot() или sns.lmplot(), необходимо передать lowess=True.

```python
# Сгенерируем искусственные данные со сложной нелинейной зависимостью
rng = np.random.default_rng(seed=42)
x = np.linspace(0, 10, 100)
y = np.sin(x) + rng.normal(0, 0.3, 100)
```

```python
# LOWESS регрессия
plt.figure(figsize=(8, 6))
sns.regplot(x=x, y=y, lowess=True, color='skyblue')

plt.title('LOWESS регрессия')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
#Пример (с использованием sns.lmplot()):
# Сгенерируем искусственные данные для примера с hue
rng = np.random.default_rng(seed=42)
x = np.linspace(-5, 5, 100)
p = 1 / (1 + np.exp(-x))
y = rng.binomial(1, p, 100)
df = pd.DataFrame({'x': x, 'y': y, 'group': rng.choice(['A', 'B'], size=100)})
```

```python
# Lmplot с логистической регрессией и разделением по hue
g = sns.lmplot(x='x', y='y', data=df, hue='group', logistic=True, palette=['crimson', 'navy'])
g.fig.suptitle('Lmplot: Логистическая регрессия с hue', y=1.02)
plt.show()
```

## 2.Матрицы Корреляций и Тепловые Карты:

### **sns.heatmap()** - Тепловая карта

это функция для визуализации данных в виде цветной матрицы. Значение каждой ячейки матрицы представляется цветом, интенсивность которого соответствует величине этого значения. Тепловые карты особенно полезны для отображения корреляционных матриц, матриц сопряженности, карт плотности и других данных, где важны относительные величины значений.

**Основные параметры:**

* data: Двумерный массив (например, DataFrame или NumPy array) с данными для тепловой карты.

* vmin и vmax: Значения, определяющие диапазон шкалы цветов. Значения вне этого диапазона будут обрезаны.
* cmap: Цветовая карта (colormap), используемая для отображения значений. Seaborn и Matplotlib предоставляют множество встроенных цветовых карт (например, 'viridis', 'magma', 'coolwarm', 'RdBu', 'Greens').
* center: Значение, относительно которого центрируется цветовая карта. Полезно для отображения отклонений от некоторого среднего значения.
* robust: Булево значение. Если True, то для определения диапазона шкалы цветов используются перцентили, а не минимальное и максимальное значения. Это делает тепловую карту менее чувствительной к выбросам.
* annot: Булево значение или массив. Если True, то значения будут отображаться в каждой ячейке. Если передан массив, то в ячейках будут отображаться значения из этого массива.
* fmt: Строка формата для отображения значений в ячейках (например, '.2f' для отображения чисел с двумя знаками после запятой).
* annot_kws: Словарь с параметрами для настройки отображения текста в ячейках (например, {'size': 10, 'color': 'white'}).
* linewidths: Ширина линий, разделяющих ячейки.
* linecolor: Цвет линий, разделяющих ячейки.
* cbar: Булево значение. Отображать ли цветовую шкалу (colorbar).
* cbar_kws: Словарь с параметрами для настройки цветовой шкалы (например, {'orientation': 'horizontal'}).
* xticklabels и yticklabels: Булево значение или список строк. Если True, то отображаются подписи на осях X и Y. Если передан список строк, то он используется в качестве подписей.
* mask: Булев массив той же формы, что и data. Если True для какой-либо ячейки, то эта ячейка не будет отображаться на тепловой карте.
* ax: Объект matplotlib.axes.Axes, на котором будет нарисована тепловая карта.

**Настройка heatmap():**

* Цветовая карта (cmap): Выбор подходящей цветовой карты очень важен для эффективного отображения данных.

 * Последовательные цветовые карты: Подходят для отображения данных, которые изменяются в одном направлении (например, от низких значений к высоким). Примеры: 'viridis', 'magma', 'Greens', 'Blues'.
 * Дивергентные цветовые карты: Подходят для отображения данных, которые отклоняются от некоторого центрального значения. Примеры: 'coolwarm', 'RdBu', 'PiYG'.
*Диапазон значений (vmin, vmax, center, robust): Настройка диапазона значений позволяет выделить интересующие области данных и улучшить контрастность тепловой карты.

* Отображение значений (annot, fmt, annot_kws): Отображение значений в ячейках может сделать тепловую карту более информативной, но может также загромоздить ее, если данных много.

* Разделительные линии (linewidths, linecolor): Разделительные линии могут помочь визуально разделить ячейки и улучшить читаемость тепловой карты.

* Цветовая шкала (cbar, cbar_kws): Настройка цветовой шкалы позволяет сделать ее более информативной и удобной для чтения.

* Маска (mask): Маска позволяет скрыть определенные области тепловой карты, чтобы выделить наиболее важные данные.

##	Стилизация и Темы Seaborn

### **sns.set_theme()** - Установка общей темы Seaborn

это функция, которая устанавливает общую тему для всех графиков Seaborn. Она позволяет настроить несколько аспектов внешнего вида графиков, таких как стиль, цветовая палитра, шрифт и т.д., одновременно. sns.set_theme() является более новым и рекомендуемым способом настройки внешнего вида графиков Seaborn по сравнению с sns.set_style(), sns.set_palette() и другими отдельными функциями

**Основные параметры:**

* context: Контекст графика. Определяет масштаб элементов графика (шрифты, линии, маркеры и т.д.). Возможные значения:
 * 'paper' (по умолчанию): Для графиков, предназначенных для публикации в научных статьях.
 * 'notebook': Для графиков, отображаемых в Jupyter Notebook или других интерактивных средах.
 * 'talk': Для графиков, используемых в презентациях.
 * 'poster': Для графиков, предназначенных для постеров.
* style: Стиль графика. Определяет общий вид графика (фон, сетка, оси и т.д.). Возможные значения:
 * 'darkgrid' (по умолчанию): Темный фон с сеткой.
 * 'whitegrid': Белый фон с сеткой.
 * 'dark': Темный фон без сетки.
 * 'white': Белый фон без сетки.
 * 'ticks': Белый фон с метками на осях.
* palette: Цветовая палитра, используемая для графиков. Может быть:
Строка с именем встроенной цветовой палитры Seaborn (например, 'deep', 'muted', 'pastel', 'bright', 'dark', 'colorblind').
Список цветов в формате HEX, RGB или RGBA.
Объект Matplotlib Colormap.
* font: Семейство шрифтов, используемое для текста на графиках.
* font_scale: Масштаб шрифта. Умножает размеры шрифтов, определенные в контексте, на этот коэффициент.
* color_codes: Булево значение. Если True, то можно использовать короткие имена цветов Seaborn (например, 'b' для синего, 'g' для зеленого, 'r' для красного).
* rc: Словарь с параметрами, которые будут переданы Matplotlib для настройки внешнего вида графиков. Это позволяет настроить любые аспекты графиков, которые не охватываются другими параметрами sns.set_theme().

### sns.set_style() - Установка стиля графика

sns.set_style() - это функция, которая устанавливает только стиль графика (фон, сетка, оси и т.д.). Она является более старой функцией и в большинстве случаев рекомендуется использовать sns.set_theme() вместо нее.

**Основные параметры:**

style: Стиль графика. Возможные значения те же, что и для sns.set_theme():
'darkgrid' (по умолчанию): Темный фон с сеткой.
'whitegrid': Белый фон с сеткой.
'dark': Темный фон без сетки.
'white': Белый фон без сетки.
'ticks': Белый фон с метками на осях.
rc: Словарь с параметрами, которые будут переданы Matplotlib для настройки внешнего вида графиков. Это позволяет настроить любые аспекты графиков, которые не охватываются параметром style.

```python
# 1. Установка стиля 'whitegrid'
sns.set_style('whitegrid')

plt.figure(figsize=(8, 6))
sns.histplot(tips['total_bill'])
plt.title('Histplot со стилем whitegrid')
plt.show()
```

```python
# 2. Установка стиля с пользовательским словарем rc
sns.set_style('dark', rc={'axes.facecolor': '#E6E6E6',
                           'grid.color': '#FFFFFF'})

plt.figure(figsize=(8, 6))
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title('Scatterplot с пользовательским стилем')
plt.show()
```

#Примеры визуализации данных о продажах

```python
# Используем pandas для загрузки данных
df = pd.read_excel('sales.xlsx', sheet_name='1',) # Укажите путь к вашему файлу
df.head()
```

```python
#Распределение количества проданных товаров:
sns.histplot(df['Quantity'], kde=True) # kde - линия плотности распределения
plt.title('Распределение количества проданных товаров')
plt.xlabel('Количество')
plt.ylabel('Частота')
plt.show()
```

**Анализ:** Этот график показывает частоту, с которой встречаются различные значения количества проданных товаров.

kde=True добавляет линию оценки плотности ядра (KDE), что помогает понять общую форму распределения.

**Закономерности:** Если гистограмма смещена вправо (большинство значений сосредоточены в левой части), это указывает на то, что большинство продаж приходится на небольшое количество товаров. Если есть несколько пиков, это может означать, что есть несколько популярных уровней количества.

```python
#Распределение цен на товары:
sns.displot(df['UnitPrice'], kde=True)
plt.title('Распределение цен на товары')
plt.xlabel('Цена')
plt.ylabel('Частота')
plt.show()
```

**Анализ:** Этот график показывает распределение цен на товары, аналогично гистограмме для количества.

**Закономерности:**Если гистограмма смещена влево, это означает, что большинство товаров имеют низкую цену. Пики в распределении могут указывать на распространенные ценовые категории.

```python
#Зависимость количества проданных товаров от цены:
sns.scatterplot(x='UnitPrice', y='Quantity', data=df)
plt.title('Зависимость количества от цены')
plt.xlabel('Цена')
plt.ylabel('Количество')
plt.show()
```

**Анализ:** Этот график показывает взаимосвязь между ценой товара и количеством проданных единиц.

**Закономерности:** Если точки образуют нисходящий тренд, это может указывать на обратную зависимость между ценой и количеством (чем выше цена, тем меньше продается). Если точки разбросаны случайно, то прямой зависимости нет. Кластеризация точек может указывать на определенные ценовые диапазоны, в которых продается больше товаров.

```python
# Средняя цена товара по категориям:
sns.barplot(x='Category', y='UnitPrice', data=df)
plt.title('Средняя цена товара по категориям')
plt.xlabel('Категория')
plt.ylabel('Средняя цена')
plt.xticks(rotation=45)  # Поворачиваем названия категорий для читаемости
plt.show()
```

**Анализ:** Этот график сравнивает средние цены товаров в разных категориях.

**Закономерности:**
Позволяет быстро определить, какие категории товаров имеют самые высокие или самые низкие средние цены.

```python
#Суммарная выручка по категориям:
df['Revenue'] = df['Quantity'] * df['UnitPrice'] # Добавляем столбец с выручкой
sns.barplot(x='Category', y='Revenue', data=df)
plt.title('Суммарная выручка по категориям')
plt.xlabel('Категория')
plt.ylabel('Выручка')
plt.xticks(rotation=45)
plt.show()
```

**Анализ:**Этот график сравнивает общую выручку, полученную от разных категорий товаров.

**Закономерности:** Помогает определить, какие категории приносят наибольший доход. Важные данные для определения приоритетов.

```python
# Распределение выручки по регионам:
sns.barplot(x='Region', y='Revenue', data=df)
plt.title('Распределение выручки по регионам')
plt.xlabel('Регион')
plt.ylabel('Выручка')
plt.show()
```

**Анализ:** Этот график показывает вклад каждого региона в общую выручку.

**Закономерности:** Позволяет сравнить эффективность продаж в разных регионах и выявить наиболее прибыльные регионы.

```python
#Ящики с усами для цен товаров по категориям:
sns.boxplot(x='Category', y='UnitPrice', data=df)
plt.title('Распределение цен по категориям (ящики с усами)')
plt.xlabel('Категория')
plt.ylabel('Цена')
plt.xticks(rotation=45)
plt.show()
```

**Закономерности:** Позволяет сравнить не только средние цены, но и разброс цен внутри каждой категории. Выбросы могут указывать на аномально дорогие или дешевые товары в категории.

```python
correlation_matrix = df.corr(numeric_only=True) # Убираем предупреждение о нечисловых типах
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm") # annot - показывать значения корреляции
plt.title("Тепловая карта корреляции признаков")
plt.show()
```
