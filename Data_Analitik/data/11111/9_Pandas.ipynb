# DataFrame

**DataFrame** — это двумерная таблица с метками (индексами строк и именами столбцов). Он обладает следующими ключевыми характеристиками:

* Столбцы могут иметь разные типы данных:  Один столбец может содержать числа, другой — строки, а третий — даты.

* Индексы строк и имена столбцов:

 Обеспечивают удобный доступ к данным.
* Возможность работы с пропущенными данными (NaN).
* Функциональность для анализа данных: Фильтрация, сортировка, группировка, агрегация и т.д.

# Манипуляции данными:

После того, как данные загружены в Pandas DataFrame, следующим шагом является их обработка и преобразование для анализа и визуализации. **Этот процесс называется манипуляцией данными.**

Он включает в себя широкий спектр операций, таких как фильтрация, сортировка, добавление и удаление столбцов, изменение типов данных, обработка пропущенных значений и многое другое.

## 1.Выбор данных (индексация и срезы)

**df[]** Выбор столбцов по имени.

**.loc[]** Выбор строк и столбцов по меткам (имена индексов и столбцов).

**.iloc[]** Выбор строк и столбцов по числовым индексам.

###  df[]  Выбор столбцов по имени.

```python
data = {'Name': ['Alice', 'Bob', 'Charlie','Liza', 'Goerg', 'Mikel','Nike '],
        'Age': [25, 30, 22, 45, 50, 18,0],
        'City': ['London', 'Paris', 'Berlin','London', 'Paris', 'Berlin','Deli'],
        'Salary': [2000, 5000, 4000, 1000,1,6000, 3000]}
df = pd.DataFrame(data)
df
```

```python
# по имени столбца
df['Name']
```

### .loc[]  Выбор строк и столбцов по меткам

```python
# по имена индексов и столбцов
print(df.loc[0])  # Выбор первой строки
print()
print(df.loc[0:1, 'Name':'Age'])  # Выбор строк 0 и 1 и столбцов 'Name' и 'Age
```

### .iloc[]: Выбор строк и столбцов по числовым индексам.

```python
print(df.iloc[0])  # Выбор первой строки
print()
print(df.iloc[0:2, 0:3])  # Выбор первых двух строк и первых двух столбцов
```

## 6.Обработка пропущенных значений (NaN)

Пропущенные значения (NaN — Not a Number) часто встречаются в данных. Pandas предоставляет методы для их обработки.

**df.isnull()** Проверка на наличие пропущенных значений (возвращает DataFrame с True/False).

**df.fillna(value)** Заполнение пропущенных значений указанным значением.

**df.dropna()** Удаление строк с пропущенными значениями.

**df.isna().sum()** просмотр пустых

Существует два подхода для работы с пропущенными значениями:
•	полное удаление строки или столбца, которые содержат пропуски
•	заполнение пропусков статистиками, например, средним значением по признаку (если мы хотим оставить строку с полезной информацией)
Обычно неинформативным считается признак, в котором пропущено 20-30% значений. Пропуски — необязательно NaN: иногда это могут быть знаки вопроса, другие символы, которые означают, что в ячейке датафрейма располагается информация, которая будет бесполезна. В этом случае нужно внимательно изучить, что означают странные значения в датафрейме, и оценить их связь со значениями в других признаках.

## 7.Агрегация и группировка данных

Агрегация позволяет вычислять статистику по группам данных.

**Наиболее часто используемые агрегатные функции:**

**'sum'** Сумма

**'mean'** Среднее

**'median'** Медиана

**'min'** Минимум

**'max'** Максимум

**'count'** Количество

**'nunique'** Количество уникальных значений


```python
# Группировка данных по "город" и вычисление среднего возраста
average_age = df.groupby("City")["Years"].mean()
average_age
```

**Группировка  groupby()**- для вычислений агрегированных данных с одновременным вычислением несколько функций

**df.groupby('column_name').agg({'column1': 'function1', 'column2': 'function2'})**

Группировка по столбцу и применение агрегатных функций.

```python
# группируем по городу .groupby('City')
# .agg()-агригируя по {года : среднее, зарплата : суммируя}
grouped = df.groupby('City').agg({'Years': 'mean', 'Salary': 'sum'})
grouped
```

```python
# Агрегирование данных с одновременным вычислением нескольких функций
grouped_data = df.groupby("City").agg({"Years": ["mean", "max"], "Salary": "sum"})
grouped_data
```

## 8.Объединение и соединение DataFrame

Pandas позволяет объединять и соединять DataFrame, как в SQL.

**pd.concat([df1, df2])** склеивает датафреймы:
*	axis=0 — присоединяет таблицу снизу;
*	axis=1 — присоединяет таблицу справа;


**pd.merge(df1, df2, on='column_name', how='join_type')** Соединение DataFrame по общему столбцу.


**Типы соединения (how):  для merge**

**'left'** Левое соединение (все строки из df1, соответствующие строки из df2).

**'right'** Правое соединение (все строки из df2, соответствующие строки из df1).

## 9.Дубликаты




Дубликаты в данных — это повторяющиеся строки, которые не приносят пользы анализу и могут искажать результаты. Удаление дубликатов — важный этап очистки данных, который позволяет получить более точные и надежные результаты.

Они могут появиться по разным причинам:

* Ошибки ввода данных: Случайное повторение строк.

* Слияние данных из разных источников: Неправильная
обработка данных из нескольких файлов или баз данных.

* Проблемы в процессе ETL: Ошибки во время извлечения, преобразования или загрузки данных.

Удаление дубликатов с помощью

**df.drop_duplicates()**


Основные параметры:
* subset,
* keep,
* inplace

###  Обнаружение дубликатов

**df.duplicated()** Возвращает Series с логическими значениями (True/False),

**df[df.duplicated()]** Отображает только строки-дубликаты.

**df.duplicated(keep='first')** По умолчанию, keep='first'. Отмечает все дубликаты, кроме первого.

**df.duplicated(keep='last')** Отмечает все дубликаты, кроме последнего.

**df.duplicated(keep=False)** Отмечает все дубликаты как True.

### Удаление дубликатов

**df.drop_duplicates()** используется для удаления дубликатов.По умолчанию, удаляет все дубликаты, сохраняя первое вхождение (keep='first').

**Основные параметры:**

* **subset:** Позволяет указать список столбцов, по которым следует искать дубликаты. Если не указан, проверяются все столбцы.

* **keep:** Определяет, какие дубликаты следует сохранить:
  * 'first' (по умолчанию): Сохраняет первое вхождение, удаляет все последующие дубликаты.

  *  'last': Сохраняет последнее вхождение, удаляет все предыдущие дубликаты.

  *  False: Удаляет все дубликаты (включая все вхождения).

* **inplace:** Если True, удаляет дубликаты непосредственно в DataFrame. Если False (по умолчанию), возвращает новый DataFrame с удаленными дубликатами.

**Обработка при наличии NaN:**

* NaN (Not a Number) представляет  собой пропущенные значения. При удалении дубликатов NaN считается равным NaN. Если у вас есть несколько строк с NaN в том же столбце, что и другие столбцы, pandas будет считать их дубликатами (в зависимости от параметров subset и keep).

```python
## Удаление дубликатов
df_no_duplicates1 = df1.drop_duplicates() # 2,4,5 полностью дублировали в маленькой
df_no_duplicates1
```

```python
## Удаление дубликатов
df_no_duplicates2 = df2.drop_duplicates() # на большой базе  2 , 4 остаются, и удаляется только 5
df_no_duplicates2
```
