# Что такое парсинг

Парсинг (из строк) => извлекает => информацию






**Основные этапы парсинга**:

**1.	Получение данных:**

*	Чтение текстовых файлов
*	Web Scraping (загрузка HTML-страниц с веб-сайтов)
*	Получение данных из API

**2.	Лексический анализ (Tokenization):**
*	Разбиение входного текста на последовательность токенов (лексем) — минимальных значимых единиц (например, слов, чисел, знаков препинания).

**3.	Синтаксический анализ (Parsing):**
*	Построение дерева разбора (parse tree) или абстрактного синтаксического дерева (AST) на основе токенов в соответствии с заданной грамматикой. Этот шаг определяет структуру данных.

**4.	Семантический анализ (Semantic Analysis):**
*	Проверка соответствия структуры данных логическим правилам и ограничениям.
*	Выявление и обработка ошибок.

**5.	Извлечение информации (Data Extraction):**
*	Выборка нужных данных из структуры, полученной на предыдущих этапах.
*	Преобразование данных в нужный формат (например, в список, словарь, объект).

**6.	Сохранение данных:**
*	Запись данных в файл (CSV, JSON, XML)
*	Сохранение в базу данных.


# BeautifulSoup: Библиотека для парсинга HTML и XML

**BeautifulSoup**— это популярная Python-библиотека, предназначенная для парсинга HTML и XML документов. Она создает дерево разбора на основе входного документа, позволяя легко перемещаться по структуре, искать элементы и извлекать данные. BeautifulSoup особенно полезна для веб-скрейпинга, когда необходимо извлекать информацию с веб-сайтов.

##Основные классы и объекты:

**BeautifulSoup:** Главный класс, представляющий весь HTML/XML документ.

**Tag:** Класс, представляющий тег `(например, <div>, <h1>, <a>)`.

Предоставляет методы для доступа к атрибутам и содержимому тега.

**NavigableString:** Класс, представляющий текст внутри тега.

**Comment:** Класс, представляющий HTML-комментарий.

## Параметры конструктора BeautifulSoup:

**markup:**

(Обязательный) HTML или XML код, который нужно распарсить. Это может быть строка, открытый файловый объект или объект Response, полученный с помощью requests.


**features:**

(Обязательный) Парсер, который будет использоваться для разбора документа. Наиболее распространенные значения:

**'html.parser':**

Встроенный парсер Python. Достаточно быстрый, но менее снисходителен к ошибкам в HTML.

**'lxml':**

Рекомендуемый парсер, если он установлен. Очень быстрый и хорошо обрабатывает невалидный HTML. Требует установки библиотеки lxml (pip install lxml).

**'xml':**

Используется для парсинга XML-документов. Требует установки lxml.

**'html5lib':**

Самый медленный, но самый снисходительный к ошибкам парсер. Требует установки библиотеки html5lib (pip install html5lib)

`soup = BeautifulSoup(html, 'lxml')`

**builder:**

 (Не рекомендуется к использованию напрямую) Этот параметр позволяет указать конкретный класс построителя дерева (например, LXML, HTMLParserTreeBuilder). Обычно лучше просто указывать features в виде строки.

**parse_only:**

Позволяет указать, какие части документа нужно парсить. Принимает объект SoupStrainer. Это может быть полезно для экономии времени и памяти, если вам нужны только определенные части документа

```
from bs4 import SoupStrainer
only_a_tags = SoupStrainer("a") # Парсим только теги <a>
soup = BeautifulSoup(html, "html.parser", parse_only=only_a_tags)
print(soup.prettify())  # Выведет только теги <a>
```


**from_encoding:**

Позволяет указать кодировку документа, если BeautifulSoup не может определить ее автоматически.

**exclude_encodings:**

Список кодировок, которые следует исключить при автоматическом определении кодировки.

**element_classes:**
 Словарь, позволяющий связать определенные теги с пользовательскими классами Python.

# Основные методы и атрибуты объектов BeautifulSoup и Tag:

## Навигация по дереву:

* .contents: Список непосредственных потомков элемента.

* .children: Итератор по непосредственным потомкам элемента.
* .parent: Непосредственный родительский элемент.
* .parents: Итератор по всем родительским элементам.
* .next_sibling: Следующий элемент на том же уровне дерева.
* .previous_sibling: Предыдущий элемент на том же уровне дерева.
* .next_siblings: Итератор по следующим элементам на том же уровне дерева.
* .previous_siblings: Итератор по предыдущим элементам на том же уровне дерева.

## Поиск элементов:

**find(name, attrs, recursive, string, **kwargs)**-  Находит первый элемент, соответствующий заданным критериям.

**name:** Имя тега (например, 'div', 'a', 'h1'). Можно передать список имен (['h1', 'h2']).

**attrs:** Словарь атрибутов и их значений (например, {'class': 'product-name', 'id': 'title'}).

**recursive:** Если True (по умолчанию), поиск ведется по всем потомкам элемента. Если False, поиск ведется только по непосредственным потомкам.

**string:** Поиск элементов, содержащих заданный текст.

****kwargs:** Позволяет передавать атрибуты как именованные аргументы (например, class_='product-name'). Обратите внимание, что class нужно указывать как class_, потому что class — это зарезервированное слово в Python.

**find_all(name, attrs, recursive, string, limit, **kwargs):** Находит все элементы, соответствующие заданным критериям.

**limit:** Ограничивает количество возвращаемых результатов.

**select(selector):** Использует CSS-селекторы для поиска элементов. Более мощный и гибкий, чем find и find_all. Требует знания CSS.

**select_one()** возвращает первый найденный элемент в виде объекта Tag (или None, если ни один элемент не соответствует селектору).

## Извлечение данных:

**.text:** Возвращает текст, содержащийся внутри элемента (включая текст всех потомков).

**.get(attribute) или ['attribute']:** Возвращает значение указанного атрибута (например, tag.get('href') или tag['href']).

**.attrs:** Возвращает словарь всех атрибутов элемента.

##Модификация дерева: (Используется реже)

**append(new_tag):** Добавляет новый тег в конец списка потомков текущего тега.

**insert(index, new_tag):** Вставляет новый тег в указанную позицию в списке потомков.

**clear():** Удаляет все содержимое тега.

**extract():** Удаляет тег из дерева и возвращает его.

**replace_with(new_tag):** Заменяет текущий тег новым тегом.

**unwrap():** Заменяет тег его содержимым.

**Вывод:**

**prettify():** Возвращает HTML-код с отступами и переносами строк, делая его более читаемым.

## примеры

````python
from bs4 import BeautifulSoup

html = """
<html>
<head><title>Пример страницы</title></head>
<body>
    <h1>Заголовок</h1>
    <p class="text">Это параграф. <a href="https://www.example.com">Ссылка</a></p>
    <ul>
        <li class="item">Элемент 1</li>
        <li class="item">Элемент 2</li>
    </ul>
</body>
</html>
"""

soup = BeautifulSoup(html, 'html.parser')

# 1. Получение заголовка страницы
title = soup.title.text  # Или soup.find('title').text
print("Заголовок:", title)

# 2. Поиск параграфа с классом "text"
paragraph = soup.find('p', class_='text') # find(): Ищет первый элемент, соответствующий селектору.
print("Параграф:", paragraph.text)

# 3. Поиск всех элементов li (списка) с классом "item"
items = soup.find_all('li', class_='item')
print("Элементы списка:")
for item in items:
    print(item.text)

# 4. Использование CSS-селекторов
link = soup.select_one('p.text a') # Находим первый тег <a> внутри <p class="text">
print("Ссылка:", link['href'])  #  прямой доступ к атрибуту через квадратные скобки
````

# рабочий пример  получение ссылок

````python
import requests
from bs4 import BeautifulSoup

# 1. Получение HTML-страницы
url = "https://supertelo906090.com/catalog/Specialnoe-Predlozhenie"
response = requests.get(url)  # получить ответ от сервера  по адресу
html = response.text     # пытается автоматически определить кодировку символов из заголовков HTTP-ответа

# 2. Создание объекта BeautifulSoup
soup = BeautifulSoup(html, 'html.parser') # Используем встроенный парсер Python

# 3. Извлечение данных
title = soup.find('title').text   # find(): Ищет первый элемент, соответствующий селектору.
print("Заголовок:", title)

# Поиск всех ссылок на странице
links = soup.find_all('a')   # find_all(): Находит все элементы, соответствующие заданным критериям.
for link in links:
    print(link.get('href'))  # Возвращает значение указанного атрибута
````

````python
import requests  # библиотека запросов
from bs4 import BeautifulSoup

def scrape_product_names(url):
    """
    Извлекает названия товаров с веб-страницы, используя указанную структуру HTML.
    Args:
        url (str): URL веб-страницы.
    Returns:
        list: Список названий товаров.  Возвращает пустой список, если произошла ошибка
              или названия не найдены.
    """
    try:
        # 1. Получение HTML-страницы
        response = requests.get(url) # получить адрес
        response.raise_for_status()  # Проверяет, не вернул ли сервер ошибку (например, 404 Not Found). Если ошибка произошла, скрипт остановится, и вы увидите сообщение об ошибке.
        html = response.text  # # пытается автоматически определить кодировку символов из заголовков HTTP-ответа


        # 2. Создание объекта BeautifulSoup
        soup = BeautifulSoup(html, 'html.parser')

        # 3. Извлечение названий товаров (надо адаптировать под структуру HTML сайта)
        product_names = []
        item_title_divs = soup.find_all('div', class_='name')  # Находим все div с нужными атрибутами

        for item_title_div in item_title_divs:
            a_tag = item_title_div.find('a', class_='prodName')  # Находим тег <a> внутри div
            if a_tag:
                product_name = a_tag.text.strip()  # Извлекаем текст  и удаляем лишние пробелы
                product_names.append(product_name)

        return product_names


    except requests.exceptions.RequestException as e:
        print(f"Ошибка при запросе страницы: {e}")
        return []
    except Exception as e:
        print(f"Ошибка при парсинге: {e}")
        return []

# Пример использования:
url = "https://supertelo906090.com/catalog/Specialnoe-Predlozhenie"  # Замените на URL вашей страницы
product_names = scrape_product_names(url)   # запускаем нашу функцию

if product_names:
    for name in product_names:
        print(f"Товар: {name}")
else:
    print("Не удалось извлечь названия товаров.")
````

## рабочий пример  вытаскивания из title тега a

````python
from bs4 import BeautifulSoup

# 1. Получение HTML-страницы
url = "https://supertelo906090.com/catalog/Specialnoe-Predlozhenie"
response = requests.get(url)  # получить ответ от сервера  по адресу
html = response.text
soup = BeautifulSoup(html, 'html.parser')

# Находим все теги <a>
links = soup.find_all('a', class_='prodName')

for link in links:
    # Получаем значение атрибута title
    title_attribute = link.get('title')  # Используем get(), чтобы избежать KeyError, если атрибут отсутствует

    # Получаем текст ссылки
    link_text = link.text.strip()  # strip() убирает лишние пробелы

    # Выводим результаты
    if title_attribute:
        print(f" Заголовок (title): {title_attribute}")
        #print(f"Текст ссылки: {link_text}, Заголовок (title): {title_attribute}")
    else:
        print(f"Текст ссылки: {link_text}, Заголовок (title): отсутствует")


# Альтернативный способ с использованием select_one и CSS-селектора
link = soup.select_one('a[title]')  # Находим первый тег <a> с атрибутом title
if link:
    title_attribute = link['title'] #или link.get('title')
    link_text = link.text.strip()
    print(f"Текст ссылки (select_one): {link_text}, Заголовок (title) (select_one): {title_attribute}")
````
