# Обработки Категориальных Признаков

## Label Encoding (Порядковое кодирование)

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
```

```python
# Создадим пример DataFrame
data = {'цвет': ['красный', 'синий', 'зеленый', 'красный', 'синий', "желтый"]}
df = pd.DataFrame(data)
df
```

```python
# Создадим экземпляр LabelEncoder
label_encoder = LabelEncoder()

# Обучим LabelEncoder на столбце 'цвет' и преобразуем его
df['цвет_encoded'] = label_encoder.fit_transform(df['цвет'])

print(df)
```

## One-Hot Encoding (Однократное кодирование)

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
```

```python
# Создадим пример DataFrame
data = {'цвет': ['красный', 'синий', 'зеленый', 'красный', 'синий', "желтый"]}
df = pd.DataFrame(data)
df
```

```python
# Создадим экземпляр OneHotEncoder
# sparse=False для возврата numpy массива, а не sparse matrix,
# handle_unknown='ignore' обрабатывает неизвестные категории,которые могут появится позже
# drop='first' удаляет один столбец для избежания мультиколлинеарности
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore', drop='first') # , drop='first')

# Обучим OneHotEncoder на столбце 'цвет' и преобразуем данные
ft= encoder.fit_transform(df[['цвет']])

# Создадим DataFrame из результатов One-Hot Encoding
onehot_df = pd.DataFrame(ft, columns=encoder.get_feature_names_out(['цвет']))

# Объединим исходный DataFrame с One-Hot Encoding
df = pd.concat([df, onehot_df], axis=1)
print(df)

# Удалим исходный столбец 'цвет'
df = df.drop('цвет', axis=1)

print(df)
```

OneHotEncoder намеренно возвращает float, чтобы обеспечить совместимость с широким спектром алгоритмов машинного обучения.

В большинстве случаев лучше оставить их как float, чтобы избежать потенциальных проблем с интерпретацией признаков алгоритмами машинного обучения

преобразовать в int можно например, для экономии памяти

```python
import numpy as np
```

```python
# можем перевести в (int) или (np.int8)

df = df.astype(np.int8)
```

### способ Dummy Variable Encoding

```python
import pandas as pd
```

```python
# Создадим пример DataFrame
data = {'цвет': ['красный', 'синий', 'зеленый', 'красный', 'синий', "желтый"]}
df = pd.DataFrame(data)
df
```

```python
# Dummy Variable Encoding с использованием pandas get_dummies()
df_dummy = pd.get_dummies(df, columns=['цвет'])
```

```python
df_dummy
```

Большинство алгоритмов машинного обучения корректно обрабатывают булевы значения как бинарные числовые значения.т.к. False и True уже представляют собой бинарные значения (0 и 1). False эквивалентно 0, а True эквивалентно 1
