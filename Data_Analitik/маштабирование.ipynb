# StandardScaler (Стандартизация)

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
# Создадим пример DataFrame
data = {'признак1': [1, 2, 3, 4, 5, 6, 7],
'признак2': [1, 2, 30, 60, 700, 900,1000]}
df = pd.DataFrame(data)
df
```

```python
# Создадим экземпляр StandardScaler
scaler = StandardScaler()

# Обучим StandardScaler на данных и преобразуем их
df_scaled = scaler.fit_transform(df)

# Преобразуем в DataFrame
df_scaled = pd.DataFrame(df_scaled, columns=df.columns)

print("Исходный DataFrame:")
print(df)
print("\nМасштабированный DataFrame (StandardScaler):")
print(df_scaled)
print("\nСтатистики после масштабирования (должны быть близки к mean=0, std=1):")
print(df_scaled.describe())
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 4))  # ncols=2 для двух графиков в строке

ax1.set_title('До StandardScaler')
sns.kdeplot(df['признак1'], ax=ax1)
sns.rugplot(df['признак1'], ax=ax1)

ax2.set_title('После StandardScaler')
sns.kdeplot(df_scaled['признак1'], ax=ax2)
sns.rugplot(df_scaled['признак1'], ax=ax2)

plt.tight_layout()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 4))  # ncols=2 для двух графиков в строке

ax1.set_title('До StandardScaler')
sns.kdeplot(df['признак2'], ax=ax1)
sns.rugplot(df['признак2'], ax=ax1)

ax2.set_title('После StandardScaler')
sns.kdeplot(df_scaled['признак2'], ax=ax2)
sns.rugplot(df_scaled['признак2'], ax=ax2)

plt.tight_layout()
plt.show()
```

# MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
```

```python
# Создадим пример DataFrame
data = {'признак1': [1, 2, 3, 4, 5, 6, 7],
'признак2': [1, 2, 30, 60, 700, 800,1000]}
df2 = pd.DataFrame(data)
df2
```

```python
# Создадим экземпляр MinMaxScaler
scaler = MinMaxScaler()

# Обучим MinMaxScaler на данных и преобразуем их
df_scaled2 = scaler.fit_transform(df2)

# Преобразуем в DataFrame
df_scaled2 = pd.DataFrame(df_scaled2, columns=df2.columns)

print("Исходный DataFrame:")
print(df2)
print("\nМасштабированный DataFrame (MinMaxScaler):")
print(df_scaled2)
print("\nСтатистики после масштабирования (должны быть в диапазоне [0, 1]):")
print(df_scaled2.describe())

```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 4))

ax1.set_title('До MinMaxScaler')
sns.kdeplot(df2['признак1'], ax=ax1)
sns.rugplot(df2['признак1'], ax=ax1)

ax2.set_title('После MinMaxScaler')
sns.kdeplot(df_scaled2['признак1'], ax=ax2)
sns.rugplot(df_scaled2['признак1'], ax=ax2)

plt.tight_layout()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 4))

ax1.set_title('До MinMaxScaler')
sns.kdeplot(df2['признак2'], ax=ax1)
sns.rugplot(df2['признак2'], ax=ax1)

ax2.set_title('После MinMaxScaler')
sns.kdeplot(df_scaled2['признак2'], ax=ax2)
sns.rugplot(df_scaled2['признак2'], ax=ax2)

plt.tight_layout()
plt.show()
```

Представьте, что у вас есть значения: [2, 4, 6, 8, 10, 100] (100 - это выброс).

Без выброса (значения: [2, 4, 6, 8, 10]): MinMaxScaler масштабирует эти значения в диапазон от 0 до 1, и они будут распределены относительно равномерно.

С выбросом (значения: [2, 4, 6, 8, 10, 100]): MinMaxScaler масштабирует эти значения в диапазон от 0 до 1, но из-за выброса значения [2, 4, 6, 8, 10] будут сконцентрированы в очень узком диапазоне около 0, а 100 станет 1.

Это “сжимает” полезную информацию в данных и делает модель более чувствительной к выбросу.
