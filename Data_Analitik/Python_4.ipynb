## Строки

Строки (strings) - это один из основных типов данных в Python, используемый для представления текста. Строка — это последовательность символов, заключенная в кавычки (одинарные '...', двойные "...", или тройные '''...''' или """...""").

**Особенности строк в Python:**

* Неизменяемость: Строки в Python являются неизменяемыми. Это означает, что после создания строки вы не можете изменить ее содержимое напрямую. Любые операции над строкой, которые кажутся изменяющими ее, на самом деле создают новую строку.

* Последовательность: Строки являются последовательностями, что означает, что вы можете получить доступ к отдельным символам в строке по их индексу (начиная с 0).

### Доступ к символам строки (индексация и срезы)

#### Индексация:

**Пример**

```python
text = "Python"
print(text[0])  # Вывод: P
print(text[1])  # Вывод: y
print(text[-1]) # Вывод: n (последний символ)
```

#### Срезы (slicing)

```python
text = "Python is fun"
print(text[0:6])   # Вывод: Python (от индекса 0 до 6, не включая 6)
print(text[7:])    # Вывод: is fun (от индекса 7 до конца строки)
print(text[:6])    # Вывод: Python (от начала строки до индекса 6, не включая 6)
print(text[::2])   # Вывод: Pto sfu (каждый второй символ)
print(text[::-1])  # Вывод: nuf si nohtyP (обратная строка)
```

### Операции над строками

#### Конкатенация (сложение)

```python
string1 = "Hello"
string2 = "World"
result = string1 + " " + string2
print(result)  # Вывод: Hello World
```

#### Повторение (умножение)

```python
string = "Python "
result = string * 3
print(result)  # Вывод: Python Python Python
```

#### Проверка наличия подстроки (есть или нет)

```python
text = "Python is fun"
print("Python" in text)  # Вывод:  True
print("Java" in text)    # Вывод: False
```

### Встроенные функции и методы строк

Python предоставляет множество встроенных функций и методов для работы со строками. Поскольку строки неизменяемы, большинство методов возвращают новую строку.

#### len(string): Возвращает длину строки (количество символов).

```python
text = "Python is, Fun"
print(len(text))  # Вывод: 14 считает с пробелами и препинаниями
```

#### Регистр:

* string.lower(): Преобразует все символы строки в нижний регистр.

* string.upper(): Преобразует все символы строки в верхний регистр.

* string.capitalize(): Преобразует первый символ строки в верхний регистр, а остальные — в нижний.

* string.title(): Преобразует первый символ каждого слова в строке в верхний регистр.

```python
text = "Python is Fun"
print(text.lower())    # Вывод: python is fun
print(text.upper())    # Вывод: PYTHON IS FUN
print(text.capitalize()) # Вывод: Python is fun
print(text.title())    # Вывод: Python Is Fun
```

#### Удаление пробелов:

* string.strip(): Удаляет начальные и конечные пробелы.

* string.lstrip(): Удаляет начальные пробелы.

* string.rstrip(): Удаляет конечные пробелы.

```python
text = " Python is fun "

print(text.strip())   # Вывод: Python is fun
print(text.lstrip())  # Вывод: Python is fun
print(text.rstrip())  # Вывод:    Python is fun
```

#### Заполнение

* string.zfill():Используется для заполнения нулями путем указания ширины поля.

```python
text = "Python is fun"
print(text.zfill(30))
```

#### Поиск и замена:

* string.find(substring): Возвращает индекс первого вхождения подстроки substring в строке. Если подстрока не найдена, возвращает -1.

* string.replace(old, new): Заменяет все вхождения подстроки old на подстроку new.

```python
# поиск
text = "Python is fun"
print(text.find("is"))    # Вывод: 7 выводит  индекс
print(text.find("Java"))  # Вывод: -1

# замена
print(text.replace("fun", "awesome"))  # Вывод: Python is awesome
```

#### Разбиение и объединение:

* string.split(separator): Разбивает строку на список подстрок, используя separator в качестве разделителя. Если separator не указан, используется пробел.

* separator.join(list): Объединяет элементы списка в одну строку, используя separator в качестве разделителя между элементами.

```python
text = "Python is fun"
words = text.split()
print(words)  # Вывод: ['Python', 'is', 'fun']

text = "apple,banana,cherry"
fruits = text.split(",")
print(fruits) # Вывод: ['apple', 'banana', 'cherry']

separator = " : "
result = separator.join(fruits)
print(result) # Вывод: apple - banana - cherry


result_2 = " * ".join(fruits)
print(result_2) # Вывод: apple - banana - cherry
```

#### Форматирование строк:

* f-strings (начиная с Python 3.6): Самый современный и удобный способ форматирования строк.

* str.format(): Более старый, но все еще широко используемый способ форматирования.

```python
# f-strings
name = "Alice"
age = 30
print(f"Это имя {name} и возраст {age} years old.")
# Вывод: My name is Alice and I am 30 years old.
```

```python
# str.format()
name = "Alice"
age = 30
print("My name is {} and I am {} years old.".format(name, age))
# Вывод: My name is Alice and I am 30 years old.

print("My name is {0} and I am {1} years old.".format(name, age)) # Явное указание индексов
# Вывод: My name is Alice and I am 30 years old.

print("My name is {name} and I am {age} years old.".format(name=name, age=age)) # Использование именованных аргументов
# Вывод: My name is Alice and I am 30 years old.
```

## Обработка ошибок (Exceptions)





Обработка ошибок (или исключений) - позволяет программе корректно реагировать на неожиданные ситуации и предотвращать аварийное завершение. В Python для обработки ошибок используются конструкции try, except, else и finally.

**Обработка исключений**

При работе с файлами важно обрабатывать возможные исключения, такие как
* FileNotFoundError (файл не найден)
* IOError (ошибка ввода/вывода).

 Конструкция try...except позволяет перехватывать исключения и обрабатывать их:

**Исключение (exception)** — это событие, которое происходит во время выполнения программы и нарушает нормальный ход ее выполнения. Исключения возникают, когда Python сталкивается с ситуацией, которую он не может обработать обычным способом,например:

* Попытка разделить число на ноль.
* Попытка открыть файл, который не существует.
* Попытка получить доступ к несуществующему индексу в списке.
* Попытка использовать переменную, которая не была определена.
* Ошибка при преобразовании типа данных.

**Примеры распространенных исключений:**

* ValueError: Некорректное значение аргумента.
* TypeError: Несоответствующий тип данных.
* IndexError: Индекс находится вне диапазона.
* KeyError: Ключ не найден в словаре.
* FileNotFoundError: Файл не найден.
* IOError: Ошибка ввода/вывода.
* ImportError: Не удалось импортировать модуль.
* NameError: Переменная не определена.
* ZeroDivisionError: Деление на ноль.

**Когда использовать обработку исключений?**

* Когда ваш код может столкнуться с неожиданными ситуациями (например, пользователь вводит некорректные данные, файл отсутствует, сетевое соединение прервано).
* Когда вам нужно гарантировать, что определенные действия будут выполнены в любом случае (например, закрытие файла).
* Когда вы хотите отделить логику обработки ошибок от основного кода программы, чтобы сделать код более читаемым и поддерживаемым.

Рекомендации по обработке исключений:

* Старайтесь обрабатывать только те исключения, которые вы можете реально обработать.
* Не перехватывайте исключения, чтобы просто их проигнорировать. Это может скрыть важные ошибки.
* Используйте finally для выполнения действий, которые должны быть выполнены в любом случае.
* Создавайте свои собственные исключения для более специфичных ситуаций в вашем приложении.
* Используйте логирование для записи информации об исключениях для последующего анализа.

#### Конструкция **try...except**

**try: **Блок кода, в котором может произойти исключение.

**except:** Блок кода, который выполняется, если в блоке try произошло исключение указанного типа.

```python
# Код, который может вызвать исключение выдаст ошибку
result = 10 / 0  # Попытка деления на ноль
print(result)
```

```python
try:
    # Код, который может вызвать исключение
    result = 10 / 0  # Попытка деления на ноль
    print(result)
except ZeroDivisionError:
    # Код, который выполняется, если произошло исключение ZeroDivisionError
    print("Ошибка: Деление на ноль!")
```

##### Обработка нескольких типов исключений

Можно обрабатывать несколько типов исключений в одном блоке try, указывая разные блоки except для каждого типа исключения:

```python
try:
    num = int(input("Введите число: "))
    result = 10 / num
    print(result)
except ValueError:
    print("Ошибка: Введено некорректное число!") # сообщение на текст
except ZeroDivisionError:
    print("Ошибка: Деление на ноль!") # сообщение на 0
```

пример  файл не найден и  ошибка ввода

```python
try:
    with open("nonexistent_file.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Файл не найден!")  # сообщение если нет файла
except IOError:
    print("Ошибка ввода/вывода!")
```

##### Блок except без указания типа исключения

Можно использовать блок except без указания типа исключения. В этом случае он будет перехватывать любое исключение, которое произойдет в блоке try. Не рекомендуется использовать это без необходимости, так как это может затруднить отладку:

```python
try:
        # Код, который может вызвать исключение
        result = 10 / 0
except:
    # Код, который выполняется, если произошло какое-либо исключение
    print("Произошла ошибка!")
```

##### Блок else

Блок else выполняется только в том случае, если в блоке try не произошло никаких исключений:

```python
try:
        num = int(input("Введите число: "))
        result = 10 / num
except ValueError:
    print("Ошибка: Введено некорректное число!")
except ZeroDivisionError:
    print("Ошибка: Деление на ноль!")
else:
    print("Результат:", result)  # Выполнится, только если не было ошибок
```

##### Блок finally

Блок finally выполняется всегда, независимо от того, произошло исключение в блоке try или нет. Он обычно используется для выполнения действий, которые необходимо выполнить в любом случае, например:

* Закрытие файла.
* Освобождение ресурсов.
* Завершение соединения с базой данных.

```python
try:
    file = open("my_file.txt", "r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("Файл не найден!")
finally:
    if 'file' in locals() and file: # Проверяем, был ли файл открыт и существует ли переменная file
        file.close()
        print("Файл закрыт.")
```

Комплексного использования try...except...else...finally:

```python
def divide(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        print("Деление на ноль!")
        return None  # Важно вернуть что-то, чтобы не получить ошибку при использовании результата
    else:
        print("Деление прошло успешно.")
        return result
    finally:
        print("Выполняется блок finally.")

print(divide(10, 2))
# Вывод:
# Деление прошло успешно.
# Выполняется блок finally.
# 5.0

print(divide(10, 0))
# Вывод:
# Деление на ноль!
# Выполняется блок finally.
# None
```

#### Генерация исключений

Вы можете сами создавать исключения с помощью ключевого слова raise. Это полезно, когда вам нужно сообщить об ошибке, которая не является стандартным исключением

в этом примере пробуем вбить возраст положительный отрицательный ,слишком высокий (свыше 120), или просто текст




1 вариант без raise

```python
# 1 вариант без генератора исключения выдает ошибку
def check_age(age):
    if age < 0:
        raise ValueError("Возраст не может быть отрицательным!")
    elif age > 120:
        raise ValueError("Возраст выглядит подозрительно высоким!")
    else:
        print("Возраст корректен.")

age = int(input("Введите ваш возраст: "))
check_age(age)
```

```python
# 2 вариант с генератором исключений
def check_age(age):
    if age < 0:
        raise ValueError("Возраст не может быть отрицательным!")
    elif age > 120:
        raise ValueError("Возраст выглядит подозрительно высоким!")
    else:
        print("Возраст корректен.")

try:
    age = int(input("Введите ваш возраст: "))
    check_age(age)
except ValueError as e: # если текст
    print("Ошибка:", e) # Выводим сообщение, переданное при генерации исключения
```

## Работа с файлами

Работа с файлами —  читать данные из файлов и записывать данные в файлы.

**Основные понятия:**

* **Файл**: Именованное место хранения данных на диске.
* **Путь к файлу**: Строка, указывающая расположение файла в файловой системе.
* **Режим доступа**: Определяет, какие операции можно выполнять с файлом (чтение, запись, добавление и т.д.).
* **Дескриптор файла** (file handle): Объект, представляющий открытый файл в программе.

**Важные советы:**

* Всегда закрывайте файлы после использования. Использование конструкции with — лучший способ гарантировать это.

* Обрабатывайте исключения, чтобы программа корректно обрабатывала ошибки при работе с файлами.

* Будьте внимательны при выборе режима доступа к файлу, чтобы избежать случайной перезаписи данных.

* Помните, что строки в Python содержат символы новой строки \n в конце каждой строки. При чтении файлов часто требуется удалять эти символы с помощью line.strip().

* При работе с путями к файлам используйте модуль os.path для обеспечения переносимости между разными операционными системами.

* **Бинарные файлы:** Файлы, содержащие нетекстовые данные (например, изображения, аудио, видео, исполняемые файлы). Используйте режимы "rb" (чтение) и "wb" (запись). При работе с бинарными файлами не используйте кодирование (encoding).

**Примеры режимов:**

"rt" (или просто "r"): Чтение текстового файла.

"wt" (или просто "w"): Запись в текстовый файл.

"rb": Чтение бинарного файла.

"wb": Запись в бинарный файл.

"r+": Чтение и запись (файл должен существовать).

"w+": Чтение и запись (файл будет перезаписан, если существует, или создан).

3. Использование библиотеки os.path для работы с путями (кросс-платформенность):

Библиотека os.path предоставляет функции для манипулирования путями к файлам, которые работают одинаково на разных операционных системах.

**Возможные записи**

* **os.path.abspath(path)**: Преобразует относительный путь в абсолютный.

* **os.path.join(path1, path2, ...)**: Объединяет несколько частей пути в один, используя правильный разделитель для текущей операционной системы (/ на macOS/Linux и \ на Windows). Это важно для кросс-платформенности.

* **os.path.split(path)**: Разделяет путь на две части: путь к каталогу и имя файла.

* **os.path.exists(path)**: Проверяет, существует ли файл или каталог по указанному пути.

* **os.path.isfile(path)**: Проверяет, является ли указанный путь файлом.

* **os.path.isdir(path)**: Проверяет, является ли указанный путь каталогом (папкой).


**Важно:**

* Разделители путей: В Windows используется обратный слэш (\), а в macOS и Linux — прямой слэш (/).

    * Для кросс-платформенности рекомендуется использовать os.path.join() для создания путей, так как он автоматически использует правильный разделитель.

    * Если вы используете прямые слэши в Windows, Python обычно их понимает, но лучше использовать os.path.join(). Обратные слэши в строках Python нужно экранировать, удваивая их (\\).

* Текущий рабочий каталог: Понимание текущего рабочего каталога важно при использовании относительных путей. Вы можете получить текущий рабочий каталог с помощью os.getcwd().

* Обработка исключений: Обязательно обрабатывайте исключения FileNotFoundError при работе с файлами, чтобы предотвратить сбой программы, если файл не найден.

##### **Просмотр содержимое архива** метод .namelist()


Просматриваем содержимое архива с помощью метода .namelist()

Метод .namelist() возвращает список имен файлов и каталогов, содержащихся в архиве

```python
import zipfile

try:
    with zipfile.ZipFile('my_archive.zip', 'r') as my_zip:
        print("Содержимое архива:")
        for name in my_zip.namelist():   # возвращает список имен  файлов и каталогов
            print(name)
except FileNotFoundError:
    print("Ошибка: Файл 'my_archive.zip' не найден.")
except zipfile.BadZipFile:
    print("Ошибка: Файл не является корректным ZIP-архивом.")
```

**my_zip.namelist():** Возвращает список имен файлов и каталогов в архиве.

**for name in ...:** Цикл для перебора имен.
print(name): Выводит каждое имя.

##### **Получение информации о размере** с помощью метода .getinfo().

Пример программы, которая выводит размер каждого файла внутри архива.

```python
import zipfile

try:
    with zipfile.ZipFile('my_archive.zip', 'r') as my_zip:
        print("Содержимое архива (с размерами):")
        for info in my_zip.infolist():  # Используем infolist() для получения информации
            file_name = info.filename
            file_size = info.file_size  # Получаем размер файла в байтах
            print(f"  {file_name}: {file_size} байт")
except FileNotFoundError:
    print("Ошибка: Файл 'my_archive.zip' не найден.")
except zipfile.BadZipFile:
    print("Ошибка: Файл не является корректным ZIP-архивом.")
```

**my_zip.infolist():** Этот метод возвращает список объектов ZipInfo. Каждый объект ZipInfo содержит информацию о файле или каталоге внутри архива. Это важнее, чем namelist(), потому что он дает нам доступ к метаданным, таким как размер.

**file_name = info.filename:** Извлекаем имя файла из объекта ZipInfo.

**file_size = info.file_size:** Извлекаем размер файла в байтах из объекта ZipInfo.

```python
#Чтение содержимого файла и подсчет количества строк:

def count_lines(filename):
    """Подсчитывает количество строк в файле."""
    count = 0
    with open(filename, "r") as file: # открываем файл
        for line in file:   # проход по строкам
            count += 1      # счетчик
    return count

filename = "my_file.txt"
line_count = count_lines(filename)
print(f"Количество строк в файле {filename}: {line_count}")
```

```python
#Копирование содержимого одного файла в другой:

def copy_file(source_filename, destination_filename):
    """Копирует содержимое одного файла в другой."""
    try:
        with open(source_filename, "r") as source_file, open(destination_filename, "w") as destination_file:
            for line in source_file:
                destination_file.write(line)
        print(f"Файл {source_filename} успешно скопирован в {destination_filename}")
    except FileNotFoundError:
        print(f"Ошибка: Файл {source_filename} не найден.")
    except Exception as e:
        print(f"Произошла ошибка: {e}")


source_filename = "my_file0.txt"
destination_filename = "my_file_copy.txt"
copy_file(source_filename, destination_filename)
```

Получение имени и расширения файла

Функция split_filename принимает имя файла.
Разделяем имя файла на части, используя точку (.) как разделитель.
Если точек больше одной, значит, есть расширение. Соединяем все части, кроме последней, чтобы получить имя файла без расширения. Последняя часть - это расширение.
Если точек нет, возвращаем исходное имя файла и пустую строку для расширения.
Возвращаем кортеж (имя, расширение).

```python
def split_filename(filename):
    """
    Разделяет имя файла на имя без расширения и расширение.
    """
    parts = filename.split(".")  # Разделяем имя файла на части, используя точку (.) как разделитель.
    if len(parts) > 1:
        name = ".".join(parts[:-1])  # Он создает новый список, который содержит все элементы списка parts, кроме последнего элемента.
        extension = parts[-1] # последний элемент
        return name, extension
    else:
        return filename, ""  # Если расширение отсутствует
```

Часто при построении моделей машинного обучения на текстах одним из входных параметров модели является частотность вхождения слова в текст. В этой задаче мы решим проблему нахождения частоты слова в документе.

В переменной document_text задан текст. Нужно найти частоту использования каждого уникального слова и положить его в словарь word_frequency. Преобразовывать слова в начальную форму не нужно.

Рекомендации:

Для удаления лишних символов используйте метод replace для текстов.
Для подсчёта частот используйте метод count на списках.

```python
document_text = 'There are the Food Courts in the cities that seats a lot of people and caters for every taste with dishes from all around the world. \
 The aromatic smells that come from the Food Courts arc so delicious that your mouth will water. \
 In recent years foreign foods have become a regular part of our life.'
```

```python
# Функция для подсчета слов
def count_dict(items):
    d = {}
    for i in items:
        if i in d:
            d[i] = d[i]+1
        else:
            d[i] = 1
    return d
```

```python
# 1)Удаление лишних символов методом replace
document_text.replace('.',' ')
document_text.replace('\\' , ' ')

# 2)разделили по пробелу
list_text = document_text.split()

# 3) создаем словарь и запустили функцию подсчета
word_frequency = dict(count_dict(list_text)) # преобразовали  в словарь
word_frequency
```
