```python
# Вклад в банке составляет X рублей. 
# Ежегодно он увеличивается на P процентов, после чего дробная часть копеек отбрасывается. 
# Определите, через сколько лет вклад составит не менее Y рублей.
# Что нужно сделать
# Напишите программу, которая по данным числам X, Y, P определяет, сколько лет пройдёт, 
# прежде чем сумма достигнет значения Y.
deposit = 50 #int(input('Вклад в банке?'))
percent = 30 #int(input('Проценты?'))
limit = 100 #int(input('Порог вклада?'))
year = 0
while deposit < limit:
    year += 1
    print(year, 'год.', deposit, '+', percent,'% =', (deposit + int(deposit * (percent/100))))
    deposit += (int(deposit * (percent/100)))
print('Кол-во лет для достижения порога:', year)
```

```python
seq_num = int(input('Введите количество чисел: '))
count = 0  # счетчик ПРОСТЫХ чисел

while seq_num > 0:
    number = int(input('Введите число: '))
    is_prime = True
    if number < 2:  # числа 0 и 1 не простые
        is_prime = False
    else:
        for divider in range(2, number):
            if number % divider == 0:
                is_prime = False  # нашли делитель - число составное
                break  # можно прервать цикл, дальше проверять не нужно
    
    if is_prime:
        count += 1  # увеличиваем счетчик простых чисел
    
    seq_num -= 1

print('Количество простых чисел в последовательности:', count)
```

Рассмотрим пример. Допустим, нам нужно найти НОД чисел 56 и 42.

Делим 56 на 42:
56 / 42 = 1 (остаток 14).
Заменяем число 56 на полученный остаток. Теперь работаем с числами 42 и 14.
Делим 42 на 14:
42 / 14 = 3 (остаток 0).
Остаток равен нулю — значит, мы нашли НОД. Это 14, так как на последнем шаге деление было без остатка.
Следовательно, по алгоритму Евклида НОД чисел 56 и 42 — это 14.
Другой пример. Возьмём числа 30 и 18:
Делим 30 на 18:
30 / 18 = 1 (остаток 12).
Теперь работаем с числами 18 и 12.
Делим 18 на 12:
18 / 12 = 1 (остаток 6).
Теперь у нас остались числа 12 и 6.
Делим 12 на 6:
12 / 6 = 2 (остаток 0).
Остаток стал 0, значит, НОД — это 6, так как последнее ненулевое число равно 6.
Таким образом, наибольший общий делитель чисел 30 и 18 — это 6.
Теперь попробуйте применить этот алгоритм для решения задачи. Последовательно делите большее число на меньшее, пока один из остатков не станет нулём.

```python
'''Задача 1. Опять налоги
У правительства одной из стран есть бухгалтерская программа, которая суммирует налоги её граждан, компаний плюс НДС с товаров. Страна развивалась, суммарные налоги увеличивались, и бухгалтеры заметили, что после добавления к общей сумме налогов некого НДС от какого-то продукта общая сумма перестала меняться… 

Нужно помочь бухгалтерам: напишите функцию, на вход которой подаются два числа — общая сумма налога tax и новый налог new_tax, который нужно добавить к общей сумме. Функция должна проверять, изменится ли показатель степени e при сложении двух чисел.

Пример 1:

Введите бюджет страны: 1.2e-12
Новые поступления (налог): 1.2e1
Результат: Бюджет увеличится
Пример 2:

Введите бюджет страны: 1.23e12
Новые поступления (налог): 1.2e1
Результат: Бюджет не изменится'''
import math
def check_tax(tax, new_tax):
    if math.isclose(tax, tax + new_tax):
        print('Бюджет не изменится')
    else:
        print('Бюджет увеличится')
tax = float(input('Введите бюджет страны: '))
new_tax = float(input('Новые поступления (налог): '))
check_tax(tax, new_tax)
```

```python
'''Задача 2. Сравнение
Так как в Python операции с вещественными числами могут давать неожиданные результаты (в частности, 0.1 + 0.2 не будет в точности равняться 0.3), стоит задача с этим как-то справляться. 

Напишите функцию eqv, которая принимает три числа и затем сравнивает сумму первых двух чисел с третьим с определённой степенью точности: до 15-го знака после точки. Если равенство выполняется, то функция возвращает True, иначе возвращает False.

Пример 1:

Введите первое число: 1.1
Введите второе число: 2.2
Введите третье число: 3.3
True

Пример 2:

Введите первое число: 1e-14
Введите второе число: 1e-14
Введите третье число: 3e-14
False'''
import math
def eqv(num1, num2, num3):
    return math.isclose(num1 + num2, num3, abs_tol=1e-15)
num1 = float(input('Введите первое число: '))
num2 = float(input('Введите второе число: '))
num3 = float(input('Введите третье число: '))
print(eqv(num1, num2, num3))
```

```python
def count_numbers(N):
    count = 0
    while N > 0:
        N //= 10
        count += 1
    return count

def change_number(number):
    digits_count = count_numbers(number)
    last_digit = number % 10
    first_digit = number // 10 ** (digits_count - 1)
    between_digits = number % 10 ** (digits_count - 1) // 10
    return last_digit * 10 ** (digits_count - 1) + between_digits * 10 + first_digit


def main():
    first_number = int(input('Введите первое число: '))
    second_number = int(input('Введите второе число: '))

    if count_numbers(first_number) < 3:
        print('В первом числе меньше трёх цифр.')
        return

    if count_numbers(second_number) < 4:
        print('Во втором числе меньше четырёх цифр.')
        return

    changed_first_number = change_number(first_number)
    changed_second_number = change_number(second_number)

    print('Изменённое первое число:', changed_first_number)
    print('Изменённое второе число:', changed_second_number)
    print('Сумма чисел:', changed_first_number + changed_second_number)


main()
```

```python
# Задача «Яйца»
def danger(depth):
    return depth ** 3 - 3 * depth ** 2 - 12 * depth + 10

max_danger = float(input('Введите максимально допустимый уровень опасности: '))
while max_danger <= 0:
    print('Ошибка: допустимый уровень опасности должен быть больше нуля.')
    max_danger = float(input('Введите максимально допустимый уровень опасности: '))

def depth_calculation(max_danger):
    left = 0
    right = 4

    while abs(danger((left + right) / 2)) > max_danger:
        middle = (left + right) / 2

        if danger(left) * danger(middle) <= 0:
            right = middle
        else:
            left = middle
    return (left + right) / 2

result = depth_calculation(max_danger)

print(f'Приблизительная глубина безопасной кладки: {result:.2f} метров')
```

```python
def danger(d_min, d_max):
    d_middle = (d_min + d_max) / 2
    middle_danger = d_middle ** 3 - 3 * d_middle ** 2 - 12 * d_middle + 10
    return d_middle, middle_danger

def depth_calculation(user_danger_level):
    lower_bound = 0
    upper_bound = 4

    while True:
        mid, danger_level = danger(lower_bound, upper_bound)

        if danger_level > 0:
            lower_bound = mid
        else:
            upper_bound = mid

        if abs(danger_level) < user_danger_level:
            return mid

max_danger = float(input("Введите максимально допустимый уровень опасности: "))
result = depth_calculation(max_danger)
print("Приблизительная глубина безопасной кладки: ", result)
```

```python
N = int(input('Кол-во участников: '))
K = int(input('Кол-во человек в команде: '))

while N % K != 0:
    print(f'{N} участников невозможно поделить на команды по {K} человек!')
    N = int(input('Кол-во участников: '))
    K = int(input('Кол-во человек в команде: '))

total = []
num = 1
for i in range(N // K):
    total.append(list(range(num, num+K)))
    num+=K

print(total)
```

```python
goods = [["яблоки", 50], ["апельсины", 190], ["груши", 100], ["нектарины", 200], ["бананы", 77]]
new_goods = []
new_goods_wch = []
new_fruit = input('Новый фрукт: ')
new_price = int(input('Цена: '))
new_goods.append(goods+(list([new_fruit, new_price])))
for item in new_goods:
    [item[0], item[1]] = [item[0], item[1]+(item[1]*0.08)]
    new_goods_wch[item]=[item[0], item[1]]
    
print(f'новый фрукт: {new_fruit}')
print(f'цена: {new_price}')
print(f'текущий ассортимент: {goods}')
print(f'новый ассортимент: {new_goods}')
```

```python
goods = [["яблоки", 50], ["апельсины", 190], ["груши", 100], ["нектарины", 200], ["бананы", 77]]

new_fruit = input('Новый фрукт: ')
new_price = int(input('Цена: '))

new_goods = goods.copy()
new_goods.append([new_fruit, new_price])

new_goods_wch = []
for item in new_goods:
    temp = item[1] * 1.08
    new_goods_wch.append([item[0], temp])

print(f'новый фрукт: {new_fruit}')
print(f'цена: {new_price}')
print(f'текущий ассортимент: {goods}')
print(f'новый ассортимент: {new_goods_wch}')
```
